<h1 style="color:DodgerBlue">Сложное наследование</h1>

Сложное наследование в C# отличается от простого тем, что оно позволяет создавать более глубокие иерархии классов, где один класс может наследоваться от другого, который уже является производным классом. Это позволяет еще больше абстрагировать общие свойства и методы, делая код более модульным и легким для поддержки.

Давайте рассмотрим пример сложного наследования, используя классы `Vehicle`, `Car`, и `Motorcycle`, а также введем новый класс `SportsCar`, который будет наследоваться от `Car`.

### Шаг 1: Базовый класс Vehicle

Начнем с определения базового класса `Vehicle`, как и в предыдущем примере.

```csharp
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }

    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}
```

### Шаг 2: Производный класс Car

Теперь определим класс `Car`, который наследуется от `Vehicle`.

```csharp
public class Car : Vehicle
{
    public int Doors { get; set; }

    public Car(string make, string model, int doors) : base(make, model)
    {
        Doors = doors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}");
    }
}
```

### Шаг 3: Производный класс SportsCar

Введем класс `SportsCar`, который наследуется от `Car`, демонстрируя сложное наследование.

```csharp
public class SportsCar : Car
{
    public bool IsConvertible { get; set; }

    public SportsCar(string make, string model, int doors, bool isConvertible) 
        : base(make, model, doors)
    {
        IsConvertible = isConvertible;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Автомобиль с откидным верхом: {IsConvertible}");
    }
}
```

### Шаг 4: Производный класс Motorcycle

Определим класс `Motorcycle`, который также наследуется от `Vehicle`.

```csharp
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }

    public Motorcycle(string make, string model, string engineType) : base(make, model)
    {
        EngineType = engineType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
    }
}
```

### Шаг 5: Использование классов

Теперь мы можем создать экземпляры наших классов и использовать их.

```csharp
Car myCar = new Car("Toyota", "Corolla", 4);
myCar.DisplayInfo();

SportsCar mySportsCar = new SportsCar("Ferrari", "488 Spider", 2, true);
mySportsCar.DisplayInfo();

Motorcycle myMotorcycle = new Motorcycle("Ducati", "Panigale V4", "V4");
myMotorcycle.DisplayInfo();
```

### Объяснение кода

- **Базовый класс `Vehicle`** предоставляет основные свойства и методы для всех транспортных средств.
- **Класс `Car`** наследует от `Vehicle` и добавляет свойство `Doors`.
- **Класс `SportsCar`** наследует от `Car`, демонстрируя сложное наследование, и добавляет свойство `IsConvertible`.
- **Класс `Motorcycle`** также наследует от `Vehicle`, но имеет свое собственное свойство `EngineType`.

### Отличие сложного наследования от простого

- **Глубина иерархии**: Сложное наследование позволяет создавать более глубокие иерархии классов, где классы могут наследоваться друг от друга на несколько уровней вниз.
- **Переиспользование кода**: Сложное наследование увеличивает возможности для переиспользования кода, поскольку классы могут наследовать не только непосредственно от базового класса, но и от других производных классов.
- **Флексибельность**: Сложное наследование обеспечивает большую гибкость в проектировании программ, позволяя разработчикам точно определять отношения между классами и их роли в системе.

Таким образом, сложное наследование в C# представляет собой мощный инструмент для создания хорошо структурированных и масштабируемых приложений, позволяя эффективно организовывать код и избегать дублирования.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке:</h4>

----

````
using System;

// ==================== Шаг 1: Базовый класс Vehicle ====================
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }

    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}

// ==================== Шаг 2: Производный класс Car ====================
public class Car : Vehicle
{
    public int Doors { get; set; }

    public Car(string make, string model, int doors) : base(make, model)
    {
        Doors = doors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}");
    }
}

// ==================== Шаг 3: Производный класс SportsCar (сложное наследование) ====================
public class SportsCar : Car
{
    public bool IsConvertible { get; set; }

    public SportsCar(string make, string model, int doors, bool isConvertible)
        : base(make, model, doors)
    {
        IsConvertible = isConvertible;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Автомобиль с откидным верхом: {IsConvertible}");
    }
}

// ==================== Шаг 4: Производный класс Motorcycle ====================
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }

    public Motorcycle(string make, string model, string engineType) : base(make, model)
    {
        EngineType = engineType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
    }
}

// ==================== Шаг 5: Использование классов (top-level код) ====================

Console.WriteLine("=== Car ===\n");
Car myCar = new Car("Toyota", "Corolla", 4);
myCar.DisplayInfo();

Console.WriteLine("\n=== SportsCar (сложное наследование: SportsCar → Car → Vehicle) ===\n");
SportsCar mySportsCar = new SportsCar("Ferrari", "488 Spider", 2, true);
mySportsCar.DisplayInfo();

Console.WriteLine("\n=== Motorcycle ===\n");
Motorcycle myMotorcycle = new Motorcycle("Ducati", "Panigale V4", "V4");
myMotorcycle.DisplayInfo();

Console.WriteLine("\n=== Полиморфизм: Vehicle[] ===\n");
Vehicle[] vehicles = { myCar, mySportsCar, myMotorcycle };
foreach (Vehicle v in vehicles)
{
    v.DisplayInfo();
    Console.WriteLine();
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также сложное наследование.

````
using System;

// ==================== Базовый класс Animal ====================
public class Animal
{
    // -------- Приватные поля --------
    private string _name;
    private int _age;
    private double _weight;

    // -------- Свойства с геттерами и сеттерами --------
    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым!");
            _name = value;
        }
    }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
            _age = value;
        }
    }

    public double Weight
    {
        get { return _weight; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("Вес должен быть положительным!");
            _weight = value;
        }
    }

    // -------- Статический атрибут --------
    public static int TotalAnimals { get; private set; } = 0;

    // ==================== Конструкторы ====================
    public Animal()
    {
        TotalAnimals++;
        _name = "Безымянное";
        _age = 0;
        _weight = 1.0;
    }

    public Animal(string name, int age, double weight) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    // ==================== Методы ====================
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт неопределённый звук.");
    }

    public void Eat(double amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Некорректное количество еды.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"{Name} поел. Новый вес: {Weight:F2} кг");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"Общее количество животных: {TotalAnimals}");
    }
}

// ==================== Производный класс Mammal (млекопитающее) ====================
public class Mammal : Animal
{
    private string _furColor;

    public string FurColor
    {
        get { return _furColor; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Цвет шерсти не может быть пустым!");
            _furColor = value;
        }
    }

    public Mammal(string name, int age, double weight, string furColor)
        : base(name, age, weight)
    {
        FurColor = furColor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Класс: Млекопитающее, Цвет шерсти: {FurColor}");
    }

    public void FeedMilk()
    {
        Console.WriteLine($"{Name} кормит детёныша молоком.");
    }
}

// ==================== Производный класс Dog (от Mammal) ====================
public class Dog : Mammal
{
    private string _breed;

    public string Breed
    {
        get { return _breed; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Порода не может быть пустой!");
            _breed = value;
        }
    }

    public Dog(string name, int age, double weight, string furColor, string breed)
        : base(name, age, weight, furColor)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Вид: Собака, Порода: {Breed}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    public void BringSlippers()
    {
        Console.WriteLine($"{Name} принёс тапочки!");
    }
}

// ==================== Производный класс GuideDog (сложное наследование: GuideDog → Dog → Mammal → Animal) ====================
public class GuideDog : Dog
{
    private string _ownerName;
    private bool _isWorking;

    public string OwnerName
    {
        get { return _ownerName; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя владельца не может быть пустым!");
            _ownerName = value;
        }
    }

    public bool IsWorking
    {
        get { return _isWorking; }
        set { _isWorking = value; }
    }

    public GuideDog(string name, int age, double weight, string furColor,
                    string breed, string ownerName, bool isWorking)
        : base(name, age, weight, furColor, breed)
    {
        OwnerName = ownerName;
        IsWorking = isWorking;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Вид: Собака-поводырь, Владелец: {OwnerName}, На работе: {IsWorking}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} (поводырь) тихо гавкает: Гав!");
    }

    public void Guide()
    {
        if (IsWorking)
            Console.WriteLine($"{Name} ведёт {OwnerName}.");
        else
            Console.WriteLine($"{Name} отдыхает.");
    }
}

// ==================== Производный класс Cat (от Mammal) ====================
public class Cat : Mammal
{
    private bool _isIndoor;
    private int _livesLeft;

    public bool IsIndoor
    {
        get { return _isIndoor; }
        set { _isIndoor = value; }
    }

    public int LivesLeft
    {
        get { return _livesLeft; }
        set
        {
            if (value < 0 || value > 9)
                throw new ArgumentOutOfRangeException("Жизней у кота от 0 до 9!");
            _livesLeft = value;
        }
    }

    public Cat(string name, int age, double weight, string furColor, bool isIndoor)
        : base(name, age, weight, furColor)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Вид: Кошка, Домашняя: {IsIndoor}, Жизней: {LivesLeft}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет...");
    }
}

// ==================== Производный класс Bird (птица) ====================
public class Bird : Animal
{
    private double _wingSpan;
    private bool _canFly;

    public double WingSpan
    {
        get { return _wingSpan; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("Размах крыльев должен быть положительным!");
            _wingSpan = value;
        }
    }

    public bool CanFly
    {
        get { return _canFly; }
        set { _canFly = value; }
    }

    public Bird(string name, int age, double weight, double wingSpan, bool canFly)
        : base(name, age, weight)
    {
        WingSpan = wingSpan;
        CanFly = canFly;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Класс: Птица, Размах крыльев: {WingSpan} м, Умеет летать: {CanFly}");
    }

    public virtual void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летает!");
        else
            Console.WriteLine($"{Name} не умеет летать.");
    }
}

// ==================== Производный класс Parrot (сложное наследование: Parrot → Bird → Animal) ====================
public class Parrot : Bird
{
    private string _phrase;

    public string Phrase
    {
        get { return _phrase; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Фраза не может быть пустой!");
            _phrase = value;
        }
    }

    public Parrot(string name, int age, double weight, double wingSpan,
                  bool canFly, string phrase)
        : base(name, age, weight, wingSpan, canFly)
    {
        Phrase = phrase;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Вид: Попугай, Фраза: {Phrase}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: {Phrase}!");
    }

    public override void Fly()
    {
        base.Fly();
        Console.WriteLine($"{Name} красиво машет крыльями.");
    }

    public void Repeat(string words)
    {
        Console.WriteLine($"{Name} повторяет: {words}");
    }
}

// ==================== Демонстрация (top-level код) ====================

Console.WriteLine("=== Иерархия классов ===");
Console.WriteLine("Animal");
Console.WriteLine("├── Mammal");
Console.WriteLine("│   ├── Dog");
Console.WriteLine("│   │   └── GuideDog   ← сложное наследование (4 уровня)");
Console.WriteLine("│   └── Cat");
Console.WriteLine("└── Bird");
Console.WriteLine("    └── Parrot         ← сложное наследование (3 уровня)");
Console.WriteLine();

Console.WriteLine("=== Создание экземпляров ===\n");

Animal genericAnimal = new Animal("Зверь", 3, 10.5);
Mammal mammal = new Mammal("Млекопитающее", 2, 15.0, "Коричневый");
Dog dog = new Dog("Рекс", 5, 25.0, "Рыжий", "Лабрадор");
GuideDog guideDog = new GuideDog("Бадди", 4, 30.0, "Золотистый", "Лабрадор", "Иван", true);
Cat cat = new Cat("Мурка", 2, 4.2, "Серый", true);
Bird bird = new Bird("Птица", 1, 0.3, 0.5, true);
Parrot parrot = new Parrot("Кеша", 1, 0.4, 0.3, true, "Кеша хороший");

Console.WriteLine("\n=== DisplayInfo через каждый класс ===\n");

genericAnimal.DisplayInfo();
Console.WriteLine();
mammal.DisplayInfo();
Console.WriteLine();
dog.DisplayInfo();
Console.WriteLine();
guideDog.DisplayInfo();
Console.WriteLine();
cat.DisplayInfo();
Console.WriteLine();
bird.DisplayInfo();
Console.WriteLine();
parrot.DisplayInfo();

Console.WriteLine("\n=== Полиморфизм: массив Animal[] ===\n");
Animal[] animals = { genericAnimal, mammal, dog, guideDog, cat, bird, parrot };
foreach (Animal a in animals)
{
    a.MakeSound();
}

Console.WriteLine("\n=== Специфичные методы классов ===\n");
dog.BringSlippers();
cat.Purr();
guideDog.Guide();
parrot.Fly();
parrot.Repeat("Привет!");

Console.WriteLine("\n=== Проверка валидации сеттеров ===\n");
try { cat.LivesLeft = 15; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { dog.Name = ""; }
catch (ArgumentException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { guideDog.OwnerName = ""; }
catch (ArgumentException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { bird.WingSpan = -1; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

Console.WriteLine("\n=== Статические члены ===\n");
Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals (через свойство): {Animal.TotalAnimals}");
````
