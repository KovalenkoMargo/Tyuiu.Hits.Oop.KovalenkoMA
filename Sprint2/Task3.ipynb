```
using System;

public abstract class Animal
{       

    public string Name { get ; set; }

    public int Age { get ; set; }

    public double Weight { get ; set; }

    public string Habitat { get ; set; }

    public static int TotalAnimals { get; private set; } = 0;

    protected Animal()
    {
    TotalAnimals++;
    Name = "Безымянное";
    Age = 0;
    Weight = 1.0;
    Habitat = "Неизвестно";
    }

    protected Animal(string name, int age, double weight, string habitat) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Среда: {Habitat}");
    }

    public void Eat(double amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество еды.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    // метод-сон восстанавливает вес
    public void Sleep(int hours)
    {
        if (hours <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество часов.");
            return;
        }
        Weight += hours * 0.01;
        Console.WriteLine($"[Animal] {Name} спал {hours} ч. Вес: {Weight:F2} кг");
    }

    //метод-миграция меняет среду обитания
    public void Migrate(string newHabitat)
    {
        Console.WriteLine($"[Animal] {Name} мигрирует: {Habitat} → {newHabitat}");
        Habitat = newHabitat;
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }

    public void Meet(Animal other)
    {
        if (other == null)
        {
            Console.WriteLine("[Animal] Не с кем знакомиться.");
            return;
        }
        Console.WriteLine($"[Animal] {Name} знакомится с {other.Name}.");
        MakeSound();
        other.MakeSound();
    }

    //метод взаимодействия — покормить другого
    public void Feed(Animal other, double amount)
    {
        if (other == null)
        {
            Console.WriteLine("[Animal] Некого кормить.");
            return;
        }
        Console.WriteLine($"[Animal] {Name} кормит {other.Name}.");
        other.Eat(amount);
    }
}




public interface IPet
{
    string Owner { get; set; }
    void ShowAffection();
}

public interface IFlyable
{
    bool CanFly { get; set; }
    void Fly();
}

public interface ISwimmable
{
    void Swim();
}

public interface IHunter
{
    void Hunt();
}





public class Mammal : Animal
{             
    public string FurColor { get ; set; }

    public int LegsCount { get ; set; }

    public Mammal(string name, int age, double weight, string habitat,
                  string furColor, int legsCount)
        : base(name, age, weight, habitat)
    {
        FurColor = furColor;
        LegsCount = legsCount;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Mammal] Цвет шерсти: {FurColor}, Ног: {LegsCount}");
    }

    //кормить молоком
    public void FeedMilk()
    {
        Console.WriteLine($"[Mammal] {Name} кормит детёныша молоком.");
    }

    //груминг
    public void Groom()
    {
        Console.WriteLine($"[Mammal] {Name} вычёсывает шерсть ({FurColor}).");
    }

    //ходить
    public void Walk()
    {
        Console.WriteLine($"[Mammal] {Name} ходит на {LegsCount} ногах.");
    }
}


public class Bird : Animal, IFlyable, IPet
{  
    public string Species { get ; set; } //вид

    public double Wingspan { get ; set; } //размах крыльев

    public string FeatherColor { get ; set; } //цвет перьев

    public bool CanFly { get; set; }

    public string Owner { get; set; }

    public Bird(string name, int age, double weight, string habitat,
                string species, double wingspan, string featherColor, bool canFly)
        : base(name, age, weight, habitat)
    {
        Species = species;
        Wingspan = wingspan;
        FeatherColor = featherColor;
        CanFly = canFly;
        Owner = "Без хозяина";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Вид: {Species}, Размах: {Wingspan} м, Перья: {FeatherColor}, Летает: {CanFly}");
    }

    public virtual void Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает с размахом {Wingspan} м!");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }

    public void ShowAffection()
    {
        Console.WriteLine($"[Bird] {Name} сидит на плече у {Owner}.");
    }

    // петь
    public void Sing()
    {
        Console.WriteLine($"[Bird] {Name} поёт красивым {FeatherColor} оперением.");
    }

    //клевать корм
    public void Peck()
    {
        Console.WriteLine($"[Bird] {Name} клюёт корм.");
    }

    // вить гнездо
    public void BuildNest()
    {
        Console.WriteLine($"[Bird] {Name} вьёт гнездо в среде «{Habitat}».");
    }
}







public class Dog : Mammal, IPet, IHunter
{
    public string Breed { get ; set; } //порода

    public int EnergyLevel { get ; set; }

    public string Owner { get; set; }

    public Dog(string name, int age, double weight, string habitat,
               string furColor, int legsCount, string breed, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        Breed = breed;
        EnergyLevel = 100;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}, Хозяин: {Owner}");
    }

    public void ShowAffection()
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} виляет хвостом перед {Owner}. Энергия: {EnergyLevel}");
    }

    public void Hunt()
    {
        EnergyLevel -= 20;
        Console.WriteLine($"[Dog] {Name} охотится! Энергия: {EnergyLevel}");
    }

    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }

    public void Bark()
    {
        Console.WriteLine($"[Dog] {Name} громко лает: ГАВ-ГАВ-ГАВ!");
    }

    public void GuardHouse()
    {
        EnergyLevel -= 15;
        Console.WriteLine($"[Dog] {Name} охраняет дом {Owner}. Энергия: {EnergyLevel}");
    }
}


public class GuideDog : Dog
{
    public string OwnerName { get ; set; } //имя владельца

    public bool IsWorking { get ; set; } 

    public string HarnessColor { get ; set; } //цвет упряжки

    public GuideDog(string name, int age, double weight, string habitat,
                    string furColor, int legsCount, string breed, string owner,
                    string ownerName, bool isWorking, string harnessColor)
        : base(name, age, weight, habitat, furColor, legsCount, breed, owner)
    {
        OwnerName = ownerName;
        IsWorking = isWorking;
        HarnessColor = harnessColor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[GuideDog] Владелец: {OwnerName}, На работе: {IsWorking}, Упряжь: {HarnessColor}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[GuideDog] {Name} (поводырь) тихо гавкает: Гав!");
    }

    public void Guide()
    {
        if (IsWorking)
            Console.WriteLine($"[GuideDog] {Name} ведёт {OwnerName}.");
        else
            Console.WriteLine($"[GuideDog] {Name} отдыхает.");
    }

    public void WearHarness()
    {
        Console.WriteLine($"[GuideDog] {Name} надевает {HarnessColor} упряжь.");
    }

    public void StopWorking()
    {
        IsWorking = false;
        Console.WriteLine($"[GuideDog] {Name} закончил работу.");
    }

    public void CrossRoad()
    {
        Console.WriteLine($"[GuideDog] {Name} переводит {OwnerName} через дорогу.");
    }
}

public class Cat : Mammal, IPet, IHunter
{
    public bool IsIndoor { get ; set; }

    public int LivesLeft { get ; set; } //жизни

    public string Owner { get; set; }

    public Cat(string name, int age, double weight, string habitat,
               string furColor, int legsCount, bool isIndoor, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}, Хозяин: {Owner}");
    }

    public void ShowAffection()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет рядом с {Owner}.");
    }

    public void Hunt()
    {
        Console.WriteLine($"[Cat] {Name} охотится на мышей!");
    }

    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }

    public void Scratch()
    {
        Console.WriteLine($"[Cat] {Name} точит когти.");
    }

    public void ClimbTree()
    {
        Console.WriteLine($"[Cat] {Name} залезла на дерево.");
    }

    public void SleepOnKeyboard()
    {
        Console.WriteLine($"[Cat] {Name} спит на клавиатуре.");
    }
}

public class Parrot : Bird
{      
    public string Phrase { get ; set; } //фраза

    public int VocabularySize { get ; set; } //слов. запас

    public Parrot(string name, int age, double weight, string habitat,
                  string species, double wingspan, string featherColor, bool canFly,
                  string phrase, int vocabularySize)
        : base(name, age, weight, habitat, species, wingspan, featherColor, canFly)
    {
        Phrase = phrase;
        VocabularySize = vocabularySize;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Parrot] Фраза: «{Phrase}», Словарный запас: {VocabularySize} слов");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Parrot] {Name} говорит: {Phrase}!");
    }

    public override void Fly()
    {
        base.Fly();
        Console.WriteLine($"[Parrot] {Name} красиво машет крыльями.");
    }

    public void Repeat(string words)
    {
        Console.WriteLine($"[Parrot] {Name} повторяет: {words}");
    }

    public void LearnWord(string word)
    {
        VocabularySize++;
        Console.WriteLine($"[Parrot] {Name} выучил слово «{word}». Словарный запас: {VocabularySize}");
    }

    public void Dance()
    {
        Console.WriteLine($"[Parrot] {Name} танцует под музыку.");
    }

    public void CrackSeed()
    {
        Console.WriteLine($"[Parrot] {Name} раскалывает семечко клювом.");
    }
}

public class Fish : Animal, ISwimmable
{        
    public string WaterType { get ; set; } //тип воды

    public int MaxDepth { get ; set; } //глубина

    public string ScaleColor { get ; set; } //цвет чешуи

    public Fish(string name, int age, double weight, string habitat,
                string waterType, int maxDepth, string scaleColor)
        : base(name, age, weight, habitat)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        ScaleColor = scaleColor;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Fish] {Name} молчит (рыбы не издают звуков).");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Fish] Вода: {WaterType}, Глубина: {MaxDepth} м, Чешуя: {ScaleColor}");
    }

    public void Swim()
    {
        Console.WriteLine($"[Fish] {Name} плавает на глубине до {MaxDepth} м в {WaterType.ToLower()} воде.");
    }

    public void Dive(int depth)
    {
        if (depth > MaxDepth)
            Console.WriteLine($"[Fish] {Name} не может нырнуть глубже {MaxDepth} м!");
        else
            Console.WriteLine($"[Fish] {Name} нырнул на {depth} м.");
    }

    public void Spawn()
    {
        Console.WriteLine($"[Fish] {Name} откладывает икру.");
    }

    public void HideInCorals()
    {
        Console.WriteLine($"[Fish] {Name} прячется в кораллах ({ScaleColor} чешуя помогает маскировке).");
    }
}


Console.WriteLine("Интерфейсы: IPet, IFlyable, ISwimmable, IHunter");
Console.WriteLine("Множественное наследование по контрактам:");
Console.WriteLine("  Dog   : Animal + Mammal + IPet + IHunter");
Console.WriteLine("  Cat   : Animal + Mammal + IPet + IHunter");
Console.WriteLine("  Bird  : Animal + IFlyable + IPet");
Console.WriteLine("  Parrot: Animal + Bird + IFlyable + IPet");
Console.WriteLine("  Fish  : Animal + ISwimmable");
Console.WriteLine();


Dog dog = new Dog("Рекс", 5, 25.0, "Дом", "Рыжий", 4, "Лабрадор", "Иван");
GuideDog guideDog = new GuideDog("Бадди", 4, 30.0, "Дом", "Золотистый", 4,
                                  "Лабрадор", "Иван", "Иван", true, "Красная");
Cat cat = new Cat("Мурка", 2, 4.2, "Дом", "Серый", 4, true, "Мария");
Bird bird = new Bird("Кеша", 1, 0.4, "Лес", "Попугай", 0.3, "Зелёный", true);
Parrot parrot = new Parrot("Гоша", 2, 0.5, "Дом", "Попугай", 0.35, "Красный",
                            true, "Гоша хороший", 15);
Fish fish = new Fish("Немо", 1, 0.2, "Океан", "Солёная", 50, "Оранжевый");
```
