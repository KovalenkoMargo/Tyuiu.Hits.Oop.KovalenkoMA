- Сложное наследование — цепочки Animal → Mammal → Dog → GuideDog (4 уровня) и Animal → Bird → Parrot (3 уровня).
- Множественное наследование — через интерфейсы IPet, IFlyable, ISwimmable, IHunter, которые классы реализуют в разных комбинациях (Dog и Cat — IPet + IHunter, Bird и Parrot — IFlyable + IPet, Fish — ISwimmable).

```
using System;

// ============================================================
// ЧАСТЬ 1. БАЗОВЫЙ АБСТРАКТНЫЙ КЛАСС Animal
// ============================================================
public abstract class Animal
{
    // -------- Приватные поля --------
    private string _name;
    private int _age;
    private double _weight;
    private string _habitat;          // НОВОЕ поле

    // -------- Свойства с геттерами и сеттерами --------
    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым!");
            _name = value;
        }
    }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
            _age = value;
        }
    }

    public double Weight
    {
        get { return _weight; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("Вес должен быть положительным!");
            _weight = value;
        }
    }

    // НОВОЕ свойство с валидацией
    public string Habitat
    {
        get { return _habitat; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Среда обитания не может быть пустой!");
            _habitat = value;
        }
    }

    // -------- Статический атрибут --------
    public static int TotalAnimals { get; private set; } = 0;

    // -------- Конструкторы --------
    protected Animal()
    {
        TotalAnimals++;
        _name = "Безымянное";
        _age = 0;
        _weight = 1.0;
        _habitat = "Неизвестно";
    }

    protected Animal(string name, int age, double weight, string habitat) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    // -------- Методы --------
    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Среда: {Habitat}");
    }

    public void Eat(double amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество еды.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    // НОВЫЙ метод — сон (восстанавливает вес)
    public void Sleep(int hours)
    {
        if (hours <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество часов.");
            return;
        }
        Weight += hours * 0.01;
        Console.WriteLine($"[Animal] {Name} спал {hours} ч. Вес: {Weight:F2} кг");
    }

    // НОВЫЙ метод — миграция (меняет среду обитания)
    public void Migrate(string newHabitat)
    {
        Console.WriteLine($"[Animal] {Name} мигрирует: {Habitat} → {newHabitat}");
        Habitat = newHabitat;
    }

    // -------- Статический метод --------
    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }

    // -------- Взаимодействие объектов --------
    public void Meet(Animal other)
    {
        if (other == null)
        {
            Console.WriteLine("[Animal] Не с кем знакомиться.");
            return;
        }
        Console.WriteLine($"[Animal] {Name} знакомится с {other.Name}.");
        MakeSound();
        other.MakeSound();
    }

    // НОВЫЙ метод взаимодействия — покормить другого
    public void Feed(Animal other, double amount)
    {
        if (other == null)
        {
            Console.WriteLine("[Animal] Некого кормить.");
            return;
        }
        Console.WriteLine($"[Animal] {Name} кормит {other.Name}.");
        other.Eat(amount);
    }
}

// ============================================================
// ЧАСТЬ 2. ИНТЕРФЕЙСЫ — для множественного наследования по контрактам
// ============================================================
public interface IPet
{
    string Owner { get; set; }
    void ShowAffection();
}

public interface IFlyable
{
    bool CanFly { get; set; }
    void Fly();
}

public interface ISwimmable
{
    void Swim();
}

public interface IHunter
{
    void Hunt();
}

// ============================================================
// ЧАСТЬ 3. ПРОМЕЖУТОЧНЫЕ КЛАССЫ (для сложного наследования)
// ============================================================

// -------- Mammal: Animal → Mammal --------
public class Mammal : Animal
{
    private string _furColor;
    private int _legsCount;              // НОВОЕ поле

    public string FurColor
    {
        get { return _furColor; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Цвет шерсти не может быть пустым!");
            _furColor = value;
        }
    }

    public int LegsCount
    {
        get { return _legsCount; }
        set
        {
            if (value < 0 || value > 4)
                throw new ArgumentOutOfRangeException("Количество ног от 0 до 4!");
            _legsCount = value;
        }
    }

    public Mammal(string name, int age, double weight, string habitat,
                  string furColor, int legsCount)
        : base(name, age, weight, habitat)
    {
        FurColor = furColor;
        LegsCount = legsCount;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Mammal] Цвет шерсти: {FurColor}, Ног: {LegsCount}");
    }

    // НОВЫЙ метод — кормить молоком
    public void FeedMilk()
    {
        Console.WriteLine($"[Mammal] {Name} кормит детёныша молоком.");
    }

    // НОВЫЙ метод — груминг
    public void Groom()
    {
        Console.WriteLine($"[Mammal] {Name} вычёсывает шерсть ({FurColor}).");
    }

    // НОВЫЙ метод — ходить
    public void Walk()
    {
        Console.WriteLine($"[Mammal] {Name} ходит на {LegsCount} ногах.");
    }
}

// -------- Bird: Animal → Bird --------
public class Bird : Animal, IFlyable, IPet
{
    private string _species;
    private double _wingspan;
    private string _featherColor;        // НОВОЕ поле

    public string Species
    {
        get { return _species; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Вид не может быть пустым!");
            _species = value;
        }
    }

    public double Wingspan
    {
        get { return _wingspan; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("Размах крыльев должен быть положительным!");
            _wingspan = value;
        }
    }

    public string FeatherColor
    {
        get { return _featherColor; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Цвет перьев не может быть пустым!");
            _featherColor = value;
        }
    }

    // Реализация IFlyable
    public bool CanFly { get; set; }

    // Реализация IPet
    public string Owner { get; set; }

    public Bird(string name, int age, double weight, string habitat,
                string species, double wingspan, string featherColor, bool canFly)
        : base(name, age, weight, habitat)
    {
        Species = species;
        Wingspan = wingspan;
        FeatherColor = featherColor;
        CanFly = canFly;
        Owner = "Без хозяина";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Вид: {Species}, Размах: {Wingspan} м, Перья: {FeatherColor}, Летает: {CanFly}");
    }

    // Реализация IFlyable
    public virtual void Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает с размахом {Wingspan} м!");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }

    // Реализация IPet
    public void ShowAffection()
    {
        Console.WriteLine($"[Bird] {Name} сидит на плече у {Owner}.");
    }

    // НОВЫЙ метод — петь
    public void Sing()
    {
        Console.WriteLine($"[Bird] {Name} поёт красивым {FeatherColor} оперением.");
    }

    // НОВЫЙ метод — клевать корм
    public void Peck()
    {
        Console.WriteLine($"[Bird] {Name} клюёт корм.");
    }

    // НОВЫЙ метод — вить гнездо
    public void BuildNest()
    {
        Console.WriteLine($"[Bird] {Name} вьёт гнездо в среде «{Habitat}».");
    }
}

// ============================================================
// ЧАСТЬ 4. ПРОИЗВОДНЫЕ КЛАССЫ (простое наследование)
// ============================================================

// -------- Dog: Mammal → Dog (сложное: Animal → Mammal → Dog) --------
public class Dog : Mammal, IPet, IHunter
{
    private string _breed;
    private int _energyLevel;

    public string Breed
    {
        get { return _breed; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Порода не может быть пустой!");
            _breed = value;
        }
    }

    public int EnergyLevel
    {
        get { return _energyLevel; }
        set
        {
            if (value < 0) _energyLevel = 0;
            else if (value > 100) _energyLevel = 100;
            else _energyLevel = value;
        }
    }

    // Реализация IPet
    public string Owner { get; set; }

    public Dog(string name, int age, double weight, string habitat,
               string furColor, int legsCount, string breed, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        Breed = breed;
        EnergyLevel = 100;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}, Хозяин: {Owner}");
    }

    // Реализация IPet
    public void ShowAffection()
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} виляет хвостом перед {Owner}. Энергия: {EnergyLevel}");
    }

    // Реализация IHunter
    public void Hunt()
    {
        EnergyLevel -= 20;
        Console.WriteLine($"[Dog] {Name} охотится! Энергия: {EnergyLevel}");
    }

    // НОВЫЙ метод
    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }

    // НОВЫЙ метод
    public void Bark()
    {
        Console.WriteLine($"[Dog] {Name} громко лает: ГАВ-ГАВ-ГАВ!");
    }

    // НОВЫЙ метод
    public void GuardHouse()
    {
        EnergyLevel -= 15;
        Console.WriteLine($"[Dog] {Name} охраняет дом {Owner}. Энергия: {EnergyLevel}");
    }
}

// -------- GuideDog: Dog → GuideDog (сложное: Animal → Mammal → Dog → GuideDog) --------
public class GuideDog : Dog
{
    private string _ownerName;
    private bool _isWorking;
    private string _harnessColor;        // НОВОЕ поле

    public string OwnerName
    {
        get { return _ownerName; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя владельца не может быть пустым!");
            _ownerName = value;
        }
    }

    public bool IsWorking
    {
        get { return _isWorking; }
        set { _isWorking = value; }
    }

    public string HarnessColor
    {
        get { return _harnessColor; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Цвет упряжи не может быть пустым!");
            _harnessColor = value;
        }
    }

    public GuideDog(string name, int age, double weight, string habitat,
                    string furColor, int legsCount, string breed, string owner,
                    string ownerName, bool isWorking, string harnessColor)
        : base(name, age, weight, habitat, furColor, legsCount, breed, owner)
    {
        OwnerName = ownerName;
        IsWorking = isWorking;
        HarnessColor = harnessColor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[GuideDog] Владелец: {OwnerName}, На работе: {IsWorking}, Упряжь: {HarnessColor}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[GuideDog] {Name} (поводырь) тихо гавкает: Гав!");
    }

    public void Guide()
    {
        if (IsWorking)
            Console.WriteLine($"[GuideDog] {Name} ведёт {OwnerName}.");
        else
            Console.WriteLine($"[GuideDog] {Name} отдыхает.");
    }

    // НОВЫЙ метод
    public void WearHarness()
    {
        Console.WriteLine($"[GuideDog] {Name} надевает {HarnessColor} упряжь.");
    }

    // НОВЫЙ метод
    public void StopWorking()
    {
        IsWorking = false;
        Console.WriteLine($"[GuideDog] {Name} закончил работу.");
    }

    // НОВЫЙ метод
    public void CrossRoad()
    {
        Console.WriteLine($"[GuideDog] {Name} переводит {OwnerName} через дорогу.");
    }
}

// -------- Cat: Mammal → Cat --------
public class Cat : Mammal, IPet, IHunter
{
    private bool _isIndoor;
    private int _livesLeft;

    public bool IsIndoor
    {
        get { return _isIndoor; }
        set { _isIndoor = value; }
    }

    public int LivesLeft
    {
        get { return _livesLeft; }
        set
        {
            if (value < 0 || value > 9)
                throw new ArgumentOutOfRangeException("Жизней у кота от 0 до 9!");
            _livesLeft = value;
        }
    }

    public string Owner { get; set; }

    public Cat(string name, int age, double weight, string habitat,
               string furColor, int legsCount, bool isIndoor, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}, Хозяин: {Owner}");
    }

    public void ShowAffection()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет рядом с {Owner}.");
    }

    public void Hunt()
    {
        Console.WriteLine($"[Cat] {Name} охотится на мышей!");
    }

    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }

    // НОВЫЙ метод
    public void Scratch()
    {
        Console.WriteLine($"[Cat] {Name} точит когти.");
    }

    // НОВЫЙ метод
    public void ClimbTree()
    {
        Console.WriteLine($"[Cat] {Name} залез на дерево.");
    }

    // НОВЫЙ метод
    public void SleepOnKeyboard()
    {
        Console.WriteLine($"[Cat] {Name} спит на клавиатуре.");
    }
}

// -------- Parrot: Bird → Parrot (сложное: Animal → Bird → Parrot) --------
public class Parrot : Bird
{
    private string _phrase;
    private int _vocabularySize;         // НОВОЕ поле

    public string Phrase
    {
        get { return _phrase; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Фраза не может быть пустой!");
            _phrase = value;
        }
    }

    public int VocabularySize
    {
        get { return _vocabularySize; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Словарный запас не может быть отрицательным!");
            _vocabularySize = value;
        }
    }

    public Parrot(string name, int age, double weight, string habitat,
                  string species, double wingspan, string featherColor, bool canFly,
                  string phrase, int vocabularySize)
        : base(name, age, weight, habitat, species, wingspan, featherColor, canFly)
    {
        Phrase = phrase;
        VocabularySize = vocabularySize;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Parrot] Фраза: «{Phrase}», Словарный запас: {VocabularySize} слов");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Parrot] {Name} говорит: {Phrase}!");
    }

    public override void Fly()
    {
        base.Fly();
        Console.WriteLine($"[Parrot] {Name} красиво машет крыльями.");
    }

    public void Repeat(string words)
    {
        Console.WriteLine($"[Parrot] {Name} повторяет: {words}");
    }

    // НОВЫЙ метод
    public void LearnWord(string word)
    {
        VocabularySize++;
        Console.WriteLine($"[Parrot] {Name} выучил слово «{word}». Словарный запас: {VocabularySize}");
    }

    // НОВЫЙ метод
    public void Dance()
    {
        Console.WriteLine($"[Parrot] {Name} танцует под музыку.");
    }

    // НОВЫЙ метод
    public void CrackSeed()
    {
        Console.WriteLine($"[Parrot] {Name} раскалывает семечко клювом.");
    }
}

// -------- Fish: Animal → Fish (простое) --------
public class Fish : Animal, ISwimmable
{
    private string _waterType;
    private int _maxDepth;
    private string _scaleColor;          // НОВОЕ поле

    public string WaterType
    {
        get { return _waterType; }
        set
        {
            if (value != "Пресная" && value != "Солёная")
                throw new ArgumentException("Тип воды: 'Пресная' или 'Солёная'!");
            _waterType = value;
        }
    }

    public int MaxDepth
    {
        get { return _maxDepth; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("Глубина должна быть положительной!");
            _maxDepth = value;
        }
    }

    public string ScaleColor
    {
        get { return _scaleColor; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Цвет чешуи не может быть пустым!");
            _scaleColor = value;
        }
    }

    public Fish(string name, int age, double weight, string habitat,
                string waterType, int maxDepth, string scaleColor)
        : base(name, age, weight, habitat)
    {
        WaterType = waterType;
        MaxDepth = maxDepth;
        ScaleColor = scaleColor;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Fish] {Name} молчит (рыбы не издают звуков).");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Fish] Вода: {WaterType}, Глубина: {MaxDepth} м, Чешуя: {ScaleColor}");
    }

    public void Swim()
    {
        Console.WriteLine($"[Fish] {Name} плавает на глубине до {MaxDepth} м в {WaterType.ToLower()} воде.");
    }

    // НОВЫЙ метод
    public void Dive(int depth)
    {
        if (depth > MaxDepth)
            Console.WriteLine($"[Fish] {Name} не может нырнуть глубже {MaxDepth} м!");
        else
            Console.WriteLine($"[Fish] {Name} нырнул на {depth} м.");
    }

    // НОВЫЙ метод
    public void Spawn()
    {
        Console.WriteLine($"[Fish] {Name} откладывает икру.");
    }

    // НОВЫЙ метод
    public void HideInCorals()
    {
        Console.WriteLine($"[Fish] {Name} прячется в кораллах ({ScaleColor} чешуя помогает маскировке).");
    }
}

// ============================================================
// ЧАСТЬ 5. ДЕМОНСТРАЦИЯ (top-level код)
// ============================================================

Console.WriteLine("=== ИЕРАРХИЯ КЛАССОВ ===");
Console.WriteLine("Animal (abstract)");
Console.WriteLine("├── Mammal");
Console.WriteLine("│   ├── Dog");
Console.WriteLine("│   │   └── GuideDog        ← сложное (4 уровня)");
Console.WriteLine("│   └── Cat");
Console.WriteLine("├── Bird");
Console.WriteLine("│   └── Parrot              ← сложное (3 уровня)");
Console.WriteLine("└── Fish");
Console.WriteLine();
Console.WriteLine("Интерфейсы: IPet, IFlyable, ISwimmable, IHunter");
Console.WriteLine("Множественное наследование по контрактам:");
Console.WriteLine("  Dog   : Animal + Mammal + IPet + IHunter");
Console.WriteLine("  Cat   : Animal + Mammal + IPet + IHunter");
Console.WriteLine("  Bird  : Animal + IFlyable + IPet");
Console.WriteLine("  Parrot: Animal + Bird + IFlyable + IPet");
Console.WriteLine("  Fish  : Animal + ISwimmable");
Console.WriteLine();

Console.WriteLine("=== СОЗДАНИЕ ЭКЗЕМПЛЯРОВ ===\n");

Dog dog = new Dog("Рекс", 5, 25.0, "Дом", "Рыжий", 4, "Лабрадор", "Иван");
GuideDog guideDog = new GuideDog("Бадди", 4, 30.0, "Дом", "Золотистый", 4,
                                  "Лабрадор", "Иван", "Иван", true, "Красная");
Cat cat = new Cat("Мурка", 2, 4.2, "Дом", "Серый", 4, true, "Мария");
Bird bird = new Bird("Кеша", 1, 0.4, "Лес", "Попугай", 0.3, "Зелёный", true);
Parrot parrot = new Parrot("Гоша", 2, 0.5, "Дом", "Попугай", 0.35, "Красный",
                            true, "Гоша хороший", 15);
Fish fish = new Fish("Немо", 1, 0.2, "Океан", "Солёная", 50, "Оранжевый");

Console.WriteLine("\n=== ПРОСТОЕ НАСЛЕДОВАНИЕ: DisplayInfo ===\n");
dog.DisplayInfo();      Console.WriteLine();
cat.DisplayInfo();      Console.WriteLine();
bird.DisplayInfo();     Console.WriteLine();
fish.DisplayInfo();     Console.WriteLine();

Console.WriteLine("=== СЛОЖНОЕ НАСЛЕДОВАНИЕ: GuideDog и Parrot ===\n");
guideDog.DisplayInfo(); Console.WriteLine();
parrot.DisplayInfo();   Console.WriteLine();

Console.WriteLine("=== ПОЛИМОРФИЗМ: массив Animal[] ===\n");
Animal[] animals = { dog, guideDog, cat, bird, parrot, fish };
foreach (Animal a in animals)
{
    a.MakeSound();
}

Console.WriteLine("\n=== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===\n");

// Dog — IPet + IHunter
IPet petDog = dog;
IHunter hunterDog = dog;
petDog.ShowAffection();
hunterDog.Hunt();
Console.WriteLine();

// Cat — IPet + IHunter
IPet petCat = cat;
IHunter hunterCat = cat;
petCat.ShowAffection();
hunterCat.Hunt();
Console.WriteLine();

// Bird — IFlyable + IPet
IFlyable flyableBird = bird;
IPet petBird = bird;
flyableBird.Fly();
petBird.ShowAffection();
Console.WriteLine();

// Parrot — IFlyable + IPet (через Bird)
IFlyable flyableParrot = parrot;
IPet petParrot = parrot;
flyableParrot.Fly();
petParrot.ShowAffection();
Console.WriteLine();

// Fish — ISwimmable
ISwimmable swimmableFish = fish;
swimmableFish.Swim();
Console.WriteLine();

Console.WriteLine("=== НОВЫЕ МЕТОДЫ ПРОИЗВОДНЫХ КЛАССОВ ===\n");

dog.Bark();
dog.GuardHouse();
guideDog.WearHarness();
guideDog.CrossRoad();
cat.Scratch();
cat.ClimbTree();
cat.SleepOnKeyboard();
bird.Sing();
bird.BuildNest();
parrot.LearnWord("Привет");
parrot.Dance();
fish.Dive(30);
fish.Spawn();
fish.HideInCorals();

Console.WriteLine("\n=== НОВЫЕ МЕТОДЫ БАЗОВЫХ КЛАССОВ ===\n");
dog.Sleep(8);
cat.Migrate("Тёплые края");
guideDog.FeedMilk();
guideDog.Groom();
guideDog.Walk();

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ===\n");
dog.Meet(cat);
dog.Feed(cat, 3.0);
Console.WriteLine();
parrot.Meet(bird);
parrot.Feed(bird, 1.0);

Console.WriteLine("\n=== ПРОВЕРКА ВАЛИДАЦИИ СЕТТЕРОВ ===\n");
try { cat.LivesLeft = 15; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { dog.Name = ""; }
catch (ArgumentException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { fish.WaterType = "Сладкая"; }
catch (ArgumentException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { guideDog.HarnessColor = ""; }
catch (ArgumentException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { fish.MaxDepth = -5; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

Console.WriteLine("\n=== СТАТИЧЕСКИЕ ЧЛЕНЫ ===\n");
Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals (через свойство): {Animal.TotalAnimals}");
```
