```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void AnimalActionDelegate(Animal animal);
public delegate void SoundDelegate();
public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }
    public string Color { get; set; }
    public bool IsWild { get; set; }
    public int Id { get; set; }

    public static int TotalAnimals { get; private set; } = 0;
    public delegate void AnimalStateHandler(string message);
    public event AnimalStateHandler StateChanged;

    protected Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
        Habitat = "Неизвестно";
        Color = "Неизвестный";
        IsWild = true;
        Id = TotalAnimals;
    }

    protected Animal(string name, int age, double weight, string habitat) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    protected void RaiseStateChanged(string message)
    {
        StateChanged?.Invoke(message);
    }
    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] #{Id} {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Среда: {Habitat}, Окрас: {Color}, Дикое: {IsWild}");
    }

    // перегрузка
    public void Eat(double amount)
    {
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел {amount} кг. Вес: {Weight:F2} кг");
    }

    public void Eat(string food)
    {
        Weight += 0.2;
        Console.WriteLine($"[Animal] {Name} поел «{food}». Вес: {Weight:F2} кг");
    }

    public void Eat(double amount, string food)
    {
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел {amount} кг «{food}». Вес: {Weight:F2} кг");
    }

    public void Move(int distance)
    {
        Console.WriteLine($"[Animal] {Name} перемещается на {distance} м.");
    }

    public void Move(int distance, string mode)
    {
        Console.WriteLine($"[Animal] {Name} перемещается на {distance} м способом «{mode}».");
    }

    public void Move(string direction)
    {
        Console.WriteLine($"[Animal] {Name} перемещается в направлении «{direction}».");
    }

    public void Migrate(string newHabitat)
    {
        Console.WriteLine($"[Animal] {Name} мигрирует: {Habitat} → {newHabitat}");
        Habitat = newHabitat;
    }
    public void Sleep(int hours)
    {
        Weight += hours * 0.01;
        RaiseStateChanged($"{Name} спал {hours} ч. Вес: {Weight:F2} кг");
    }

    public void Play()
    {
        RaiseStateChanged($"{Name} играет.");
    }

    public void Rest()
    {
        RaiseStateChanged($"{Name} отдыхает.");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }

    public void Meet(Animal other)
    {
        Console.WriteLine($"[Animal] {Name} знакомится с {other.Name}.");
        MakeSound();
        other.MakeSound();
    }
}




public interface IPet
{
    string Owner { get; set; }
    void ShowAffection();
}

public interface IFlyable
{
    bool CanFly { get; set; }
    void Fly();
}

public interface IHunter
{
    void Hunt();
}



public class Mammal : Animal
{
    public string FurColor { get; set; }
    public int LegsCount { get; set; }

    public bool HasTail { get; set; }
    public string DietType { get; set; }
    public double BodyTemp { get; set; }

    public Mammal(string name, int age, double weight, string habitat,
                  string furColor, int legsCount)
        : base(name, age, weight, habitat)
    {
        FurColor = furColor;
        LegsCount = legsCount;
        HasTail = true;
        DietType = "Всеядный";
        BodyTemp = 37.0;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Mammal] Шерсть: {FurColor}, Ног: {LegsCount}, Хвост: {HasTail}, Питание: {DietType}");
    }

    public void Groom()
    {
        RaiseStateChanged($"{Name} вычёсывает шерсть ({FurColor}).");
    }

    public void Groom(string brush)
    {
        RaiseStateChanged($"{Name} вычёсывает шерсть щёткой «{brush}».");
    }

    public void Walk()
    {
        RaiseStateChanged($"{Name} ходит на {LegsCount} ногах.");
    }

    public void Walk(int steps)
    {
        RaiseStateChanged($"{Name} прошёл {steps} шагов.");
    }

    public void FeedMilk()
    {
        RaiseStateChanged($"{Name} кормит детёныша молоком.");
    }

    public void Hibernate()
    {
        RaiseStateChanged($"{Name} впадает в спячку.");
    }

    public void ShedFur()
    {
        RaiseStateChanged($"{Name} линяет ({FurColor}).");
    }

    public void RegulateTemp()
    {
        RaiseStateChanged($"{Name} поддерживает t° = {BodyTemp}°C.");
    }
}

public class Bird : Animal, IFlyable, IPet
{
    public string Species { get; set; }
    public double Wingspan { get; set; }
    public string FeatherColor { get; set; }
    public bool CanFly { get; set; }
    public string Owner { get; set; }

    public int EggsPerYear { get; set; }
    public bool IsMigratory { get; set; }
    public string BeakType { get; set; }

    public Bird(string name, int age, double weight, string habitat,
                string species, double wingspan, string featherColor, bool canFly)
        : base(name, age, weight, habitat)
    {
        Species = species;
        Wingspan = wingspan;
        FeatherColor = featherColor;
        CanFly = canFly;
        Owner = "Без хозяина";
        EggsPerYear = 5;
        IsMigratory = false;
        BeakType = "Короткий";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Вид: {Species}, Размах: {Wingspan} м, Перья: {FeatherColor}, Летает: {CanFly}, Яиц/год: {EggsPerYear}");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IFlyable
    void IFlyable.Fly()
    {
        if (CanFly)
            RaiseStateChanged($"{Name} летает с размахом {Wingspan} м!");
        else
            RaiseStateChanged($"{Name} не умеет летать.");
    }

    public void FlyPublic()
    {
        ((IFlyable)this).Fly();
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IPet
    void IPet.ShowAffection()
    {
        RaiseStateChanged($"{Name} сидит на плече у {Owner}.");
    }

    public void ShowAffectionPublic()
    {
        ((IPet)this).ShowAffection();
    }

    // перегрузка
    public void Fly(int height)
    {
        if (CanFly)
            RaiseStateChanged($"{Name} летает на высоте {height} м.");
        else
            RaiseStateChanged($"{Name} не умеет летать.");
    }

    public void Sing()
    {
        RaiseStateChanged($"{Name} поёт.");
    }

    public void Sing(string song)
    {
        RaiseStateChanged($"{Name} поёт «{song}».");
    }

    public void LayEggs()
    {
        RaiseStateChanged($"{Name} откладывает яйца ({EggsPerYear} шт/год).");
    }

    public void MigrateSeason()
    {
        if (IsMigratory)
            RaiseStateChanged($"{Name} мигрирует на юг.");
        else
            RaiseStateChanged($"{Name} остаётся зимовать.");
    }

    public void Peck()
    {
        RaiseStateChanged($"{Name} клюёт корм.");
    }

    public void BuildNest()
    {
        RaiseStateChanged($"{Name} вьёт гнездо.");
    }

    public void Preen()
    {
        RaiseStateChanged($"{Name} чистит перья ({FeatherColor}).");
    }
}

public class Dog : Mammal, IPet, IHunter
{
    public string Breed { get; set; }
    public int EnergyLevel { get; set; }
    public string Owner { get; set; }

    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public int BarkVolume { get; set; }

    public Dog(string name, int age, double weight, string habitat,
               string furColor, int legsCount, string breed, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        Breed = breed;
        EnergyLevel = 100;
        Owner = owner;
        IsTrained = false;
        FavoriteToy = "Мяч";
        BarkVolume = 70;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}, Хозяин: {Owner}, Дрессирован: {IsTrained}, Игрушка: {FavoriteToy}");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IHunter
    void IHunter.Hunt()
    {
        EnergyLevel -= 20;
        RaiseStateChanged($"{Name} охотится! Энергия: {EnergyLevel}");
    }

    // НЕЯВНАЯ РЕАЛИЗАЦИЯ IPet
    public void ShowAffection()
    {
        EnergyLevel -= 5;
        RaiseStateChanged($"{Name} виляет хвостом перед {Owner}. Энергия: {EnergyLevel}");
    }

    // перегрузка
    public void Bark()
    {
        RaiseStateChanged($"{Name} лает: ГАВ!");
    }

    public void Bark(int times)
    {
        Console.Write($"[Dog] {Name} лает {times} раз: ");
        for (int i = 0; i < times; i++) Console.Write("ГАВ! ");
        Console.WriteLine();
    }

    public void Bark(int times, int volume)
    {
        BarkVolume = volume;
        RaiseStateChanged($"{Name} лает {times} раз с громкостью {volume}.");
    }

    public void BringSlippers()
    {
        EnergyLevel -= 10;
        RaiseStateChanged($"{Name} принёс тапочки! Энергия: {EnergyLevel}");
    }

    public void BringSlippers(string item)
    {
        EnergyLevel -= 5;
        RaiseStateChanged($"{Name} принёс «{item}»! Энергия: {EnergyLevel}");
    }

    public void Train()
    {
        IsTrained = true;
        RaiseStateChanged($"{Name} теперь дрессирован!");
    }

    public void PlayWithToy()
    {
        EnergyLevel -= 8;
        RaiseStateChanged($"{Name} играет с «{FavoriteToy}». Энергия: {EnergyLevel}");
    }

    public void GuardHouse()
    {
        EnergyLevel -= 15;
        RaiseStateChanged($"{Name} охраняет дом {Owner}. Энергия: {EnergyLevel}");
    }
}

public class GuideDog : Dog
{
    public string OwnerName { get; set; }
    public bool IsWorking { get; set; }
    public string HarnessColor { get; set; }

    public int YearsOfService { get; set; }
    public string Certification { get; set; }
    public bool KnowsCommands { get; set; }

    public GuideDog(string name, int age, double weight, string habitat,
                    string furColor, int legsCount, string breed, string owner,
                    string ownerName, bool isWorking, string harnessColor)
        : base(name, age, weight, habitat, furColor, legsCount, breed, owner)
    {
        OwnerName = ownerName;
        IsWorking = isWorking;
        HarnessColor = harnessColor;
        YearsOfService = 1;
        Certification = "Базовая";
        KnowsCommands = true;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[GuideDog] Владелец: {OwnerName}, На работе: {IsWorking}, Упряжь: {HarnessColor}, Лет службы: {YearsOfService}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[GuideDog] {Name} (поводырь) тихо гавкает: Гав!");
    }

    public void Guide()
    {
        if (IsWorking)
            RaiseStateChanged($"{Name} ведёт {OwnerName}.");
        else
            RaiseStateChanged($"{Name} отдыхает.");
    }

    public void WearHarness()
    {
        RaiseStateChanged($"{Name} надевает {HarnessColor} упряжь.");
    }

    public void WearHarness(string color)
    {
        HarnessColor = color;
        RaiseStateChanged($"{Name} надевает упряжь цвета «{color}».");
    }

    public void Retire()
    {
        IsWorking = false;
        RaiseStateChanged($"{Name} вышел на пенсию после {YearsOfService} лет службы.");
    }

    public void GetCertified(string level)
    {
        Certification = level;
        RaiseStateChanged($"{Name} получил сертификат «{level}».");
    }

    public void CrossRoad()
    {
        RaiseStateChanged($"{Name} переводит {OwnerName} через дорогу.");
    }

    public void StopWorking()
    {
        IsWorking = false;
        RaiseStateChanged($"{Name} закончил работу.");
    }

    public void CheckHarness()
    {
        RaiseStateChanged($"{Name} проверяет упряжь «{HarnessColor}».");
    }
}

public class Cat : Mammal, IPet, IHunter
{
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }
    public string Owner { get; set; }

    public bool IsNeutered { get; set; }
    public string FavoriteFood { get; set; }
    public int PurrVolume { get; set; }

    public Cat(string name, int age, double weight, string habitat,
               string furColor, int legsCount, bool isIndoor, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Owner = owner;
        IsNeutered = false;
        FavoriteFood = "Рыба";
        PurrVolume = 50;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}, Хозяин: {Owner}, Стерилизован: {IsNeutered}, Еда: {FavoriteFood}");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IHunter
    void IHunter.Hunt()
    {
        RaiseStateChanged($"{Name} охотится на мышей!");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IPet
    void IPet.ShowAffection()
    {
        RaiseStateChanged($"{Name} мурлычет рядом с {Owner}.");
    }

    public void ShowAffectionPublic()
    {
        ((IPet)this).ShowAffection();
    }

    // перегрузка
    public void Purr()
    {
        RaiseStateChanged($"{Name} мурлычет...");
    }

    public void Purr(int minutes)
    {
        RaiseStateChanged($"{Name} мурлычет {minutes} минут...");
    }

    public void Purr(int minutes, int volume)
    {
        PurrVolume = volume;
        RaiseStateChanged($"{Name} мурлычет {minutes} мин с громкостью {volume}.");
    }

    public void EatFavorite()
    {
        RaiseStateChanged($"{Name} ест любимую еду «{FavoriteFood}».");
    }

    public void Scratch()
    {
        RaiseStateChanged($"{Name} точит когти.");
    }

    public void ClimbTree()
    {
        RaiseStateChanged($"{Name} залезла на дерево.");
    }
}

public class Parrot : Bird
{
    public string Phrase { get; set; }
    public int VocabularySize { get; set; }

    public string BeakColor { get; set; }
    public bool CanTalk { get; set; }
    public int LifespanYears { get; set; }

    public Parrot(string name, int age, double weight, string habitat,
                  string species, double wingspan, string featherColor, bool canFly,
                  string phrase, int vocabularySize)
        : base(name, age, weight, habitat, species, wingspan, featherColor, canFly)
    {
        Phrase = phrase;
        VocabularySize = vocabularySize;
        BeakColor = "Жёлтый";
        CanTalk = true;
        LifespanYears = 50;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Parrot] Фраза: «{Phrase}», Словарь: {VocabularySize}, Клюв: {BeakColor}, Говорит: {CanTalk}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Parrot] {Name} говорит: {Phrase}!");
    }

    // перегрузка
    public void Repeat(string words)
    {
        RaiseStateChanged($"{Name} повторяет: {words}");
    }

    public void Repeat(string words, int times)
    {
        for (int i = 0; i < times; i++)
            RaiseStateChanged($"{Name} повторяет: {words}");
    }

    public void LearnWord(string word)
    {
        VocabularySize++;
        RaiseStateChanged($"{Name} выучил слово «{word}». Словарь: {VocabularySize}");
    }

    public void LearnWord(string word, int repeat)
    {
        VocabularySize += repeat;
        RaiseStateChanged($"{Name} выучил слово «{word}» ({repeat} раз). Словарь: {VocabularySize}");
    }

    public void ChangeBeakColor(string color)
    {
        BeakColor = color;
        RaiseStateChanged($"{Name} сменил цвет клюва на «{color}».");
    }

    public void ShowLifespan()
    {
        RaiseStateChanged($"{Name} может прожить до {LifespanYears} лет.");
    }

    public void Dance()
    {
        RaiseStateChanged($"{Name} танцует.");
    }
}




public class AnimalCollection<T> where T : Animal
{
    private List<T> _animals = new List<T>();

    public void Add(T animal) => _animals.Add(animal);
    public void Remove(T animal) => _animals.Remove(animal);
    public int Count => _animals.Count;

    public void DisplayAll()
    {
        foreach (var a in _animals)
        {
            a.DisplayInfo();
            a.MakeSound();
            Console.WriteLine();
        }
    }
}

public static class EventHandlers
{
    public static void OnStateChanged(string message)
    {
        Console.WriteLine($"[Событие] {message}");
    }

    public static void OnStateChangedDetailed(string message)
    {
        Console.WriteLine($"[Детально] {message} ({DateTime.Now:HH:mm:ss})");
    }
}




Dog dog = new Dog("Рекс", 5, 25.0, "Дом", "Рыжий", 4, "Лабрадор", "Иван");
GuideDog guideDog = new GuideDog("Бадди", 4, 30.0, "Дом", "Золотистый", 4, "Лабрадор", "Иван", "Иван", true, "Красная");
Cat cat = new Cat("Мурка", 2, 4.2, "Дом", "Серый", 4, true, "Мария");
Bird bird = new Bird("Кеша", 1, 0.4, "Лес", "Попугай", 0.3, "Зелёный", true);
Parrot parrot = new Parrot("Гоша", 2, 0.5, "Дом", "Попугай", 0.35, "Красный", true, "Гоша хороший", 15);
```

```
Console.WriteLine("List<Animal>\n");

List<Animal> animalList = new List<Animal> { dog, guideDog, cat, bird, parrot };

foreach (Animal a in animalList)
{
    a.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine($"Всего в List: {animalList.Count}");
```

```
Console.WriteLine("\n Dictionary<string, Animal>\n");

Dictionary<string, Animal> animalDict = new Dictionary<string, Animal>
{
    { "Рекс", dog },
    { "Бадди", guideDog },
    { "Мурка", cat },
    { "Кеша", bird },
    { "Гоша", parrot }
};

foreach (var pair in animalDict)
{
    Console.WriteLine($"{pair.Key}: {pair.Value.Name} ({pair.Value.GetType().Name})");
}
```

```
Console.WriteLine("\nHashSet<Animal>\n");
HashSet<Animal> animalSet = new HashSet<Animal>();
foreach (var a in animalList)
    animalSet.Add(a);

Console.WriteLine($"Уникальных животных: {animalSet.Count}");



Console.WriteLine("\nДелегаты\n");
// Делегат без параметров
SoundDelegate soundDog = dog.MakeSound;
SoundDelegate soundCat = cat.MakeSound;
SoundDelegate soundBird = bird.MakeSound;

soundDog();
soundCat();
soundBird();

Console.WriteLine();

AnimalActionDelegate displayAction = (a) => a.DisplayInfo();
displayAction(dog);
Console.WriteLine();
displayAction(cat);

Console.WriteLine("\n-Многоадресный делегат-");
SoundDelegate multiSound = dog.MakeSound;
multiSound += cat.MakeSound;
multiSound += bird.MakeSound;
multiSound();
```

```
Console.WriteLine("\nСобытия\n");
dog.StateChanged += EventHandlers.OnStateChanged;
guideDog.StateChanged += EventHandlers.OnStateChangedDetailed;
cat.StateChanged += EventHandlers.OnStateChanged;
bird.StateChanged += EventHandlers.OnStateChangedDetailed;
parrot.StateChanged += EventHandlers.OnStateChanged;

Console.WriteLine("-Методы собаки-");
dog.Train();
dog.PlayWithToy();
dog.GuardHouse();

Console.WriteLine("\n-Методы кошки-");
cat.EatFavorite();
cat.Scratch();
cat.ClimbTree();

Console.WriteLine("\n-Методы птицы-");
bird.LayEggs();
bird.MigrateSeason();
bird.Peck();
```

```
Console.WriteLine("\n-Методы попугая-");
parrot.Repeat("Привет");
parrot.LearnWord("Hello");
parrot.Dance();

Console.WriteLine("\n-Методы поводыря-");
guideDog.Guide();
guideDog.WearHarness("Синяя");
guideDog.CrossRoad();
```
