<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ==================== Делегат ====================
public delegate void DisplayDelegate();

// ==================== Базовый класс Transport ====================
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}

// ==================== Обработчики событий в статическом классе ====================
public static class TransportEventHandlers
{
    public static void OnTransportUpdated(string message)
    {
        Console.WriteLine($"[Событие] {message}");
    }
}

// ==================== Демонстрация (top-level код) ====================

Console.WriteLine("=== A. Делегаты ===\n");

Car car = new Car("Model S", "Tesla", 4);
Bike bike = new Bike("Speedster", "Giant", true);

DisplayDelegate displayCar = car.DisplayInfo;
DisplayDelegate displayBike = bike.DisplayInfo;

displayCar();
Console.WriteLine();
displayBike();

Console.WriteLine("\n=== B. Лямбда-выражения + LINQ ===\n");

List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
    new Bike("City", "Stels", false),
    new Car("Camry", "Toyota", 4)
};

var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var c in teslaCars)
{
    c.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine("=== Сортировка через лямбду ===\n");

transports.Sort((a, b) => a.Manufacturer.CompareTo(b.Manufacturer));
foreach (var t in transports)
{
    Console.WriteLine($"{t.Manufacturer} {t.Model}");
}

Console.WriteLine("\n=== Поиск через лямбду ===\n");

var foundCar = transports.FirstOrDefault(t => t.Model == "Camry");
if (foundCar != null)
{
    foundCar.DisplayInfo();
}

Console.WriteLine("\n=== C. События ===\n");

Transport transport = new Transport("Old Model", "Generic");

// Подписка через статический метод
transport.TransportUpdated += TransportEventHandlers.OnTransportUpdated;

transport.UpdateModel("New Model");

// Подписка через лямбду
transport.TransportUpdated += (msg) => Console.WriteLine($"[Лямбда-подписка] {msg}");
transport.UpdateModel("Newest Model");

// Отписка
transport.TransportUpdated -= TransportEventHandlers.OnTransportUpdated;
transport.UpdateModel("Final Model");
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением  делегатов, лямда-выражений и событий.

````
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void AnimalActionDelegate(Animal animal);
public delegate void SoundDelegate();

public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }

    // 3 атрибута
    public string Color { get; set; }
    public bool IsWild { get; set; }
    public int Id { get; set; }

    public static int TotalAnimals { get; private set; } = 0;

    // -------- Делегат + событие --------
    public delegate void AnimalStateHandler(string message);
    public event AnimalStateHandler StateChanged;

    // -------- Защищённый метод для вызова события --------
    // Наследники вызывают ЭТОТ метод, а не само событие
    protected void OnStateChanged(string message)
    {
        StateChanged?.Invoke(message);
    }

    protected Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
        Habitat = "Неизвестно";
        Color = "Неизвестный";
        IsWild = true;
        Id = TotalAnimals;
    }

    protected Animal(string name, int age, double weight, string habitat) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] #{Id} {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Среда: {Habitat}, Окрас: {Color}, Дикое: {IsWild}");
    }

    // Методы, вызывающие событие — внутри Animal это разрешено
    public void ChangeWeight(double newWeight)
    {
        double oldWeight = Weight;
        Weight = newWeight;
        OnStateChanged($"{Name}: вес {oldWeight:F2} → {newWeight:F2} кг");
    }

    public void ChangeHabitat(string newHabitat)
    {
        string oldHabitat = Habitat;
        Habitat = newHabitat;
        OnStateChanged($"{Name}: среда «{oldHabitat}» → «{newHabitat}»");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public int EnergyLevel { get; set; }

    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public int BarkVolume { get; set; }

    public Dog(string name, int age, double weight, string habitat, string breed)
        : base(name, age, weight, habitat)
    {
        Breed = breed;
        EnergyLevel = 100;
        IsTrained = false;
        FavoriteToy = "Мяч";
        BarkVolume = 70;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}, Дрессирован: {IsTrained}, Игрушка: {FavoriteToy}");
    }

    public void Train()
    {
        IsTrained = true;
        OnStateChanged($"{Name} теперь дрессирован!");    // ← через protected-метод
    }

    public void PlayWithToy()
    {
        EnergyLevel -= 8;
        OnStateChanged($"{Name} играет с «{FavoriteToy}». Энергия: {EnergyLevel}");
    }

    public void GuardHouse()
    {
        EnergyLevel -= 15;
        OnStateChanged($"{Name} охраняет дом. Энергия: {EnergyLevel}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }

    public bool IsNeutered { get; set; }
    public string FavoriteFood { get; set; }
    public int PurrVolume { get; set; }

    public Cat(string name, int age, double weight, string habitat, bool isIndoor)
        : base(name, age, weight, habitat)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        IsNeutered = false;
        FavoriteFood = "Рыба";
        PurrVolume = 50;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}, Стерилизован: {IsNeutered}, Еда: {FavoriteFood}");
    }

    public void EatFavorite()
    {
        OnStateChanged($"{Name} ест любимую еду «{FavoriteFood}».");
    }

    public void Scratch()
    {
        OnStateChanged($"{Name} точит когти.");
    }

    public void ClimbTree()
    {
        OnStateChanged($"{Name} залезла на дерево.");
    }
}

public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    public int EggsPerYear { get; set; }
    public bool IsMigratory { get; set; }
    public string BeakType { get; set; }

    public Bird(string name, int age, double weight, string habitat,
                double wingspan, bool canFly)
        : base(name, age, weight, habitat)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        EggsPerYear = 5;
        IsMigratory = false;
        BeakType = "Короткий";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Размах: {Wingspan} м, Летает: {CanFly}, Яиц/год: {EggsPerYear}, Перелётная: {IsMigratory}");
    }

    public void LayEggs()
    {
        OnStateChanged($"{Name} откладывает яйца ({EggsPerYear} шт/год).");
    }

    public void MigrateSeason()
    {
        if (IsMigratory)
            OnStateChanged($"{Name} мигрирует на юг.");
        else
            OnStateChanged($"{Name} остаётся зимовать.");
    }

    public void Peck()
    {
        OnStateChanged($"{Name} клюёт корм.");
    }
}

public static class AnimalEventHandlers
{
    public static void OnAnimalStateChanged(string message)
    {
        Console.WriteLine($"[Событие] {message}");
    }

    public static void OnAnimalStateChangedDetailed(string message)
    {
        Console.WriteLine($"[Детально] {message} (время: {DateTime.Now:HH:mm:ss})");
    }
}

Dog dog = new Dog("Рекс", 5, 25.0, "Дом", "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, "Дом", true);
Bird bird = new Bird("Кеша", 1, 0.4, "Лес", 0.3, true);

Console.WriteLine("Делегаты\n");

SoundDelegate soundDog = dog.MakeSound;
SoundDelegate soundCat = cat.MakeSound;
SoundDelegate soundBird = bird.MakeSound;

soundDog();
soundCat();
soundBird();

Console.WriteLine();

AnimalActionDelegate displayAction = (a) => a.DisplayInfo();

Console.WriteLine("--- Делегат с параметром ---");
displayAction(dog);
Console.WriteLine();
displayAction(cat);
Console.WriteLine();
displayAction(bird);

Console.WriteLine("\n--- Многоадресный делегат ---");
SoundDelegate multiSound = dog.MakeSound;
multiSound += cat.MakeSound;
multiSound += bird.MakeSound;
multiSound();
````

````
Console.WriteLine("\nЛямбда-выражения\n");

List<Animal> animals = new List<Animal>
{
    dog, cat, bird,
    new Dog("Шарик", 3, 15.0, "Двор", "Овчарка"),
    new Cat("Барсик", 4, 5.5, "Дом", false),
    new Bird("Попугай", 2, 0.35, "Лес", 0.25, true)
};

Console.WriteLine("--- Только собаки ---");
var dogs = animals.OfType<Dog>();
foreach (var d in dogs)
{
    Console.WriteLine($"{d.Name} ({d.Breed})");
}

Console.WriteLine("\n--- Животные весом > 5 кг ---");
var heavyAnimals = animals.Where(a => a.Weight > 5);
foreach (var a in heavyAnimals)
{
    Console.WriteLine($"{a.Name}: {a.Weight:F2} кг");
}

Console.WriteLine("\n--- Сортировка по возрасту ---");
var sortedByAge = animals.OrderBy(a => a.Age).ToList();
foreach (var a in sortedByAge)
{
    Console.WriteLine($"{a.Name}: {a.Age} лет");
}

Console.WriteLine("\n--- Проекция: только имена ---");
var names = animals.Select(a => a.Name).ToList();
Console.WriteLine(string.Join(", ", names));

Console.WriteLine("\n--- Первое животное с весом > 10 ---");
var firstHeavy = animals.FirstOrDefault(a => a.Weight > 10);
if (firstHeavy != null)
    Console.WriteLine($"{firstHeavy.Name} ({firstHeavy.Weight:F2} кг)");

Console.WriteLine("\n--- Группировка по типу ---");
var grouped = animals.GroupBy(a => a.GetType().Name);
foreach (var group in grouped)
{
    Console.WriteLine($"Тип: {group.Key} — {group.Count()} шт.");
}

Console.WriteLine("\n--- Проверка условий ---");
Console.WriteLine($"Все старше 0 лет? {animals.All(a => a.Age > 0)}");
Console.WriteLine($"Есть кто-то с весом > 20? {animals.Any(a => a.Weight > 20)}");
````

````
Console.WriteLine("\nСобытия\n");

dog.StateChanged += AnimalEventHandlers.OnAnimalStateChanged;
cat.StateChanged += (msg) => Console.WriteLine($"[Cat-событие] {msg}");
bird.StateChanged += AnimalEventHandlers.OnAnimalStateChanged;
bird.StateChanged += AnimalEventHandlers.OnAnimalStateChangedDetailed;

Console.WriteLine("--- Собака: тренировка и игры ---");
dog.Train();
dog.PlayWithToy();
dog.GuardHouse();

Console.WriteLine("\n--- Кошка: еда и когти ---");
cat.EatFavorite();
cat.Scratch();
cat.ClimbTree();

Console.WriteLine("\n--- Птица: яйца и миграция ---");
bird.LayEggs();
bird.MigrateSeason();
bird.Peck();

Console.WriteLine("\n--- Изменение состояния животных ---");
dog.ChangeWeight(27.5);
cat.ChangeHabitat("Квартира");
bird.ChangeWeight(0.45);

Console.WriteLine("\n--- Отписка от события ---");
dog.StateChanged -= AnimalEventHandlers.OnAnimalStateChanged;
dog.ChangeWeight(28.0);
````

````
Console.WriteLine("\nВсё вместе\n");

AnimalActionDelegate applyToAll = (a) =>
{
    Console.Write($"→ {a.Name}: ");
    a.MakeSound();
};

foreach (var a in animals)
{
    applyToAll(a);
}

Console.WriteLine("\n--- Подписка с фильтром по типу ---");

foreach (var a in animals)
{
    Animal captured = a;
    a.StateChanged += (msg) =>
    {
        if (captured is Dog)
            Console.WriteLine($"[Только собаки] {msg}");
    };
}

dog.PlayWithToy();
cat.Scratch();
dog.Train();

dog.Train();
cat.EatFavorite();
bird.LayEggs();

Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals: {Animal.TotalAnimals}");
````
