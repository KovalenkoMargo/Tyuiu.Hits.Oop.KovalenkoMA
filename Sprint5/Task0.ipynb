<h1 style="color:DodgerBlue">Коллекции в .NET </h1>



Коллекции в .NET - это структуры данных, которые позволяют организовать и управлять наборами объектов. Они предоставляют эффективные способы хранения, доступа и манипулирования данными. Вот основные типы коллекций в .NET:

### 1. Статические коллекции

Статические коллекции являются неизменяемыми и могут содержать только один экземпляр заданного типа данных.

- `System.Collections.Generic.EmptyCollection<T>`
- `System.Collections.Generic.Singleton<T>`
- `System.Collections.ObjectModel.ReadOnlyCollection<T>`

### 2. Неупорядоченные коллекции

Неупорядоченные коллекции не гарантируют порядок элементов.

- `System.Collections.Generic.List<T>` - динамический массив
- `System.Collections.Generic.Dictionary<TKey,TValue>` - словарь (ассоциативный массив)
- `System.Collections.Generic.HashSet<T>` - множество уникальных элементов
- `System.Collections.Generic.SortedSet<T>` - отсортированный набор элементов

### 3. Сортированные коллекции

Сортированные коллекции поддерживают упорядоченный доступ и могут быть отсортированы по определенному критерию.

- `System.Collections.Generic.SortedDictionary<TKey,TValue>`
- `System.Collections.Generic.SortedSet<T>`
- `System.Collections.Generic.SortedList<TKey,TValue>`

### 4. Множества

Множества содержат только уникальные элементы.

- `System.Collections.Generic.HashSet<T>`
- `System.Collections.Generic.ISet<T>` - интерфейс для неупорядоченных множеств

### 5. Ключевые коллекции

Ключевые коллекции используются для хранения пар ключ-значение.

- `System.Collections.Generic.Dictionary<TKey,TValue>`
- `System.Collections.Generic.KeyedCollection<TKey,TElement>`
- `System.Collections.Generic.Lookup<TKey,TValue>`

### 6. Коллекции с ограниченным размером

Эти коллекции имеют фиксированный или максимальный размер.

- `System.Collections.Generic.Stack<T>` - стек
- `System.Collections.Generic.Queue<T>` - очередь
- `System.Collections.Generic.ConcurrentQueue<T>` - потокобезопасная очередь

### 7. Потокобезопасные коллекции

Потокобезопасные коллекции обеспечивают безопасность доступа из нескольких потоков.

- `System.Collections.Concurrent.ConcurrentDictionary<TKey,TValue>`
- `System.Collections.Concurrent.ConcurrentBag<T>`
- `System.Collections.Concurrent.BlockingCollection<T>`

### 8. Специализированные коллекции

Специализированные коллекции предназначены для конкретных задач.

- `System.Collections.Generic.BCLPortabilityLibrary.Collections.Generic.MultiMap<TKey,TKey2,TValue>`
- `System.Collections.Generic.BCLPortabilityLibrary.Collections.Generic.MultiDictionary<TKey,TValue>`

Это основные типы коллекций в .NET Framework и .NET Core. Выбор конкретной коллекции зависит от требований вашего проекта, таких как необходимость сортировки, потокобезопасности или специфики использования данных.

Работа с `List<T>` и `Dictionary<TKey, TValue>` в C# позволяет эффективно управлять коллекциями объектов различных типов. Давайте рассмотрим, как это можно сделать на примере классов `Transport`, `Car`, и `Bike`.

### Класс Transport и его производные

Начнем с базового класса `Transport`, который будет основой для классов `Car` и `Bike`.

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет наситель: {HasCarrier}");
    }
}
```

### Работа с List<T>

Теперь давайте посмотрим, как работать с `List<T>` для хранения коллекции объектов транспорта.

```csharp
using System;
using System.Collections.Generic;


// Создаем список Transport
List<Transport> transportList = new List<Transport>();

// Добавляем объекты Car и Bike в список
transportList.Add(new Car("Model S", "Tesla", 4));
transportList.Add(new Bike("Mountain Explorer", "Giant", true));

// Перебираем и выводим информацию обо всех транспортных средствах
foreach (var transport in transportList)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Работа с Dictionary<TKey, TValue>

Теперь давайте создадим коллекцию, которая будет хранить объекты `Transport`, используя `Dictionary<TKey, TValue>`, где ключом будет, например, регистрационный номер транспорта.

```csharp
using System;
using System.Collections.Generic;

// Создаем словарь с ключом типа string и значением типа Transport
Dictionary<string, Transport> transportDictionary = new Dictionary<string, Transport>();

// Добавляем объекты с уникальными ключами
transportDictionary.Add("TSL123", new Car("Model S", "Tesla", 4));
transportDictionary.Add("BIK456", new Bike("Mountain Explorer", "Giant", true));

// Поиск и вывод информации о транспорте по ключу
if (transportDictionary.TryGetValue("TSL123", out Transport foundTransport))
{
    Console.WriteLine("Найденый траснпорт:");
    foundTransport.DisplayInfo();
}
else
{
    Console.WriteLine("Транспорт не найден");
}

```

### Объяснения и примеры использования:

- **List<T>**: Используется, когда порядок элементов важен и возможны дубликаты. Вы можете добавлять, удалять, итерировать и изменять элементы по индексу.

- **Dictionary<TKey,TValue>**: Подходит, когда нужно организовать элементы по уникальным ключам для быстрого доступа. Это удобно для случаев, когда вы хотите находить элемент по ключу, как регистрационный номер в примере.

Эти структуры данных предоставляют мощные и гибкие возможности для управления коллекциями объектов, каждая из которых имеет свои преимущества в определённых ситуациях.

````
using System;
using System.Collections.Generic;

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет носитель: {HasCarrier}");
    }
}

List<Transport> transportList = new List<Transport>();

transportList.Add(new Car("Model S", "Tesla", 4));
transportList.Add(new Bike("Mountain Explorer", "Giant", true));
transportList.Add(new Car("Camry", "Toyota", 4));
transportList.Add(new Bike("City Bike", "Stels", false));

foreach (var transport in transportList)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine($"Всего в списке: {transportList.Count}");

Console.WriteLine("\n=== Работа с Dictionary<string, Transport> ===\n");

Dictionary<string, Transport> transportDictionary = new Dictionary<string, Transport>();

transportDictionary.Add("TSL123", new Car("Model S", "Tesla", 4));
transportDictionary.Add("BIK456", new Bike("Mountain Explorer", "Giant", true));
transportDictionary.Add("CAM789", new Car("Camry", "Toyota", 4));

// Поиск по ключу
if (transportDictionary.TryGetValue("TSL123", out Transport foundTransport))
{
    Console.WriteLine("Найденный транспорт:");
    foundTransport.DisplayInfo();
}
else
{
    Console.WriteLine("Транспорт не найден");
}

Console.WriteLine("\n=== Перебор всех элементов словаря ===\n");

foreach (var pair in transportDictionary)
{
    Console.WriteLine($"Ключ: {pair.Key}");
    pair.Value.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine($"Всего в словаре: {transportDictionary.Count}");

Console.WriteLine("\n=== Проверка наличия ключа ===\n");

string searchKey = "BIK456";
if (transportDictionary.ContainsKey(searchKey))
{
    Console.WriteLine($"Ключ «{searchKey}» есть в словаре.");
}

Console.WriteLine("\n=== Удаление элемента ===\n");

transportDictionary.Remove("CAM789");
Console.WriteLine($"После удаления: {transportDictionary.Count} элементов.");
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте при помощи  List<T> и  Dictionary<TKey,TValue>. 

````
using System;
using System.Collections.Generic;

public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }
    public string Color { get; set; }
    public bool IsWild { get; set; }
    public int Id { get; set; }

    public static int TotalAnimals { get; private set; } = 0;
    protected Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
        Habitat = "Неизвестно";
        Color = "Неизвестный";
        IsWild = true;
        Id = TotalAnimals;
    }

    protected Animal(string name, int age, double weight, string habitat) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] #{Id} {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Среда: {Habitat}, Окрас: {Color}, Дикое: {IsWild}");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public int EnergyLevel { get; set; }

    // 3 атрибута
    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public int BarkVolume { get; set; }

    public Dog(string name, int age, double weight, string habitat,
               string breed)
        : base(name, age, weight, habitat)
    {
        Breed = breed;
        EnergyLevel = 100;
        IsTrained = false;
        FavoriteToy = "Мяч";
        BarkVolume = 70;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}, Дрессирован: {IsTrained}, Игрушка: {FavoriteToy}");
    }

    public void Train()
    {
        IsTrained = true;
        Console.WriteLine($"[Dog] {Name} теперь дрессирован!");
    }

    public void PlayWithToy()
    {
        EnergyLevel -= 8;
        Console.WriteLine($"[Dog] {Name} играет с «{FavoriteToy}». Энергия: {EnergyLevel}");
    }

    public void GuardHouse()
    {
        EnergyLevel -= 15;
        Console.WriteLine($"[Dog] {Name} охраняет дом. Энергия: {EnergyLevel}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }

    // 3 атрибута
    public bool IsNeutered { get; set; }
    public string FavoriteFood { get; set; }
    public int PurrVolume { get; set; }

    public Cat(string name, int age, double weight, string habitat,
               bool isIndoor)
        : base(name, age, weight, habitat)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        IsNeutered = false;
        FavoriteFood = "Рыба";
        PurrVolume = 50;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}, Стерилизован: {IsNeutered}, Еда: {FavoriteFood}");
    }

    // 3 новых метода
    public void EatFavorite()
    {
        Console.WriteLine($"[Cat] {Name} ест любимую еду «{FavoriteFood}».");
    }

    public void Scratch()
    {
        Console.WriteLine($"[Cat] {Name} точит когти.");
    }

    public void ClimbTree()
    {
        Console.WriteLine($"[Cat] {Name} залезла на дерево.");
    }
}

public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }

    // 3 атрибута
    public int EggsPerYear { get; set; }
    public bool IsMigratory { get; set; }
    public string BeakType { get; set; }

    public Bird(string name, int age, double weight, string habitat,
                double wingspan, bool canFly)
        : base(name, age, weight, habitat)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        EggsPerYear = 5;
        IsMigratory = false;
        BeakType = "Короткий";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Размах: {Wingspan} м, Летает: {CanFly}, Яиц/год: {EggsPerYear}, Перелётная: {IsMigratory}");
    }

    public void LayEggs()
    {
        Console.WriteLine($"[Bird] {Name} откладывает яйца ({EggsPerYear} шт/год).");
    }

    public void MigrateSeason()
    {
        if (IsMigratory)
            Console.WriteLine($"[Bird] {Name} мигрирует на юг.");
        else
            Console.WriteLine($"[Bird] {Name} остаётся зимовать.");
    }

    public void Peck()
    {
        Console.WriteLine($"[Bird] {Name} клюёт корм.");
    }
}

Dog dog = new Dog("Рекс", 5, 25.0, "Дом", "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, "Дом", true);
Bird bird = new Bird("Кеша", 1, 0.4, "Лес", 0.3, true);

Dog dog2 = new Dog("Шарик", 3, 15.0, "Двор", "Овчарка");
Cat cat2 = new Cat("Барсик", 4, 5.5, "Дом", false);
Bird bird2 = new Bird("Попугай", 2, 0.35, "Лес", 0.25, true);

Console.WriteLine("LIST<Animal>: все животные\n");

List<Animal> animalList = new List<Animal>();

animalList.Add(dog);
animalList.Add(cat);
animalList.Add(bird);
animalList.Add(dog2);
animalList.Add(cat2);
animalList.Add(bird2);

foreach (Animal a in animalList)
{
    a.DisplayInfo();
    a.MakeSound();
    Console.WriteLine();
}

Console.WriteLine($"Всего в List: {animalList.Count}");
````

````
Console.WriteLine("Поиск в List по имени\n");

string findName = "Мурка";
Animal found = animalList.Find(a => a.Name == findName);
if (found != null)
{
    Console.WriteLine($"Найден «{found.Name}»:");
    found.DisplayInfo();
}
else
{
    Console.WriteLine($"«{findName}» не найден.");
}

Console.WriteLine("\nФильтрация List: только Dog\n");

List<Animal> dogsOnly = animalList.FindAll(a => a is Dog);
foreach (Animal a in dogsOnly)
{
    a.DisplayInfo();
    Console.WriteLine();
}
Console.WriteLine($"Собак в списке: {dogsOnly.Count}");

Console.WriteLine("Удаление из List\n");

animalList.Remove(dog2);
Console.WriteLine($"После удаления Шарика: {animalList.Count}");
````

````
Console.WriteLine("\nDICTIONARY<string, Animal>: по кличке\n");

Dictionary<string, Animal> animalDict = new Dictionary<string, Animal>();

animalDict.Add("Рекс", dog);
animalDict.Add("Мурка", cat);
animalDict.Add("Кеша", bird);
animalDict.Add("Шарик", dog2);
animalDict.Add("Барсик", cat2);
animalDict.Add("Попугай", bird2);

foreach (var pair in animalDict)
{
    Console.WriteLine($"Ключ: {pair.Key}");
    pair.Value.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine($"Всего в Dictionary: {animalDict.Count}");
````

````
Console.WriteLine("Поиск в Dictionary по ключу\n");

string searchKey = "Мурка";
if (animalDict.TryGetValue(searchKey, out Animal foundAnimal))
{
    Console.WriteLine($"Найден по ключу «{searchKey}»:");
    foundAnimal.MakeSound();
    foundAnimal.DisplayInfo();
}
else
{
    Console.WriteLine($"Ключ «{searchKey}» не найден.");
}

Console.WriteLine("\nПроверка наличия ключа\n");

Console.WriteLine($"Содержит «Рекс»: {animalDict.ContainsKey("Рекс")}");
Console.WriteLine($"Содержит «Тузик»: {animalDict.ContainsKey("Тузик")}");

Console.WriteLine("\n Удаление из Dictionary\n");

animalDict.Remove("Шарик");
Console.WriteLine($"После удаления «Шарик»: {animalDict.Count} элементов.");

Console.WriteLine("\nList + Dictionary вместе\n");

// Dictionary с Id как ключом — для быстрого поиска
Dictionary<int, Animal> animalById = new Dictionary<int, Animal>();

foreach (Animal a in animalList)
{
    if (!animalById.ContainsKey(a.Id))
        animalById.Add(a.Id, a);
}
````

````
Console.WriteLine("Животные по Id:");
foreach (var pair in animalById)
{
    Console.WriteLine($"Id {pair.Key}: {pair.Value.Name} ({pair.Value.GetType().Name})");
}

Console.WriteLine("\nГруппировка по типу\n");

Dictionary<string, List<Animal>> byType = new Dictionary<string, List<Animal>>();

foreach (Animal a in animalList)
{
    string typeName = a.GetType().Name;
    if (!byType.ContainsKey(typeName))
        byType[typeName] = new List<Animal>();
    byType[typeName].Add(a);
}

foreach (var pair in byType)
{
    Console.WriteLine($"Тип: {pair.Key} — {pair.Value.Count} шт.");
    foreach (Animal a in pair.Value)
    {
        Console.WriteLine($" - {a.Name}");
    }
}

dog.Train();
dog.PlayWithToy();
dog.GuardHouse();

cat.EatFavorite();
cat.Scratch();
cat.ClimbTree();

bird.LayEggs();
bird.MigrateSeason();
bird.Peck();

Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals: {Animal.TotalAnimals}");
````
