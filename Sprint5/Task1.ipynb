<h1 style="color:DodgerBlue">HashSet<T></h1>



`HashSet<T>` в C# — это коллекция, которая хранит уникальные элементы без определенного порядка. Она полезна, когда нужно быстро проверять наличие элемента или добавлять только уникальные элементы. Давайте рассмотрим, как с ней работать на примере классов `Transport`, `Car`, и `Bike`.

### Создание классов Transport, Car, и Bike

Сначала определим базовый класс `Transport` и его наследники `Car` и `Bike`:

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        // Простая проверка равенства базируется на модели и производителе
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        // Создание хеш-кода на основе модели и производителя
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}
```

### Работа с HashSet<T>

Теперь рассмотрим, как использовать `HashSet<T>` для хранения объектов `Transport`.

```csharp
using System;
using System.Collections.Generic;


// Создаем HashSet для уникальных объектов транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Добавляем объекты Car и Bike в HashSet
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
 var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

// Попробуем добавить повторяющийся объект
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Was duplicate car added? {isAdded}"); // false

// Перебираем и выводим информацию обо всех уникальных транспортных средствах
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Объяснения и комментарии:

- **Уникальность объектов**: `HashSet<T>` не допускает дублирования. Если вы пытаетесь добавить элемент, который уже присутствует (определяется методом `Equals` и `GetHashCode`), он не добавляется.

- **Методы `Equals` и `GetHashCode`**: Эти методы переопределены для корректной работы с `HashSet`. `Equals` определяет, когда два объекта считаются равными, а `GetHashCode` возвращает хеш-код, используемый для быстрого поиска элемента.

- **Быстрая проверка на наличие**: Благодаря хешированию, `HashSet` позволяет быстро проверять, существует ли элемент в коллекции (`O(1)` — амортизированное время).

- **Отсутствие порядка**: Элементы в `HashSet` не имеют определенного порядка, в отличие от `List<T>`.

`HashSet<T>` идеально подходит для сценариев, где требуется исключительно уникальность элементов и быстрое выполнение операций поиска.

````
using System;
using System.Collections.Generic;

// ==================== Базовый класс Transport ====================
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    // Переопределяем Equals — два Transport равны, если совпадают Model и Manufacturer
    public override bool Equals(object obj)
    {
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    // Переопределяем GetHashCode — на основе Model и Manufacturer
    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}

// ==================== Car ====================
public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

// ==================== Bike ====================
public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}

// ==================== Демонстрация (top-level код) ====================

Console.WriteLine("=== Работа с HashSet<Transport> ===\n");

// Создаём HashSet для уникальных транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Добавляем объекты
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

// Пробуем добавить дубликат (та же Model + Manufacturer, что у car1)
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Дубликат добавлен? {isAdded}");   // false
Console.WriteLine();

// Перебираем и выводим информацию обо всех уникальных транспортных средствах
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

Console.WriteLine($"Всего в HashSet: {transportSet.Count}");

// -------- Проверка наличия элемента --------
Console.WriteLine("\n=== Проверка наличия элемента ===\n");

var checkCar = new Car("Model S", "Tesla", 4);   // такой же, как car1
Console.WriteLine($"Contains(car1-копия)? {transportSet.Contains(checkCar)}");   // true

var checkOther = new Car("Unknown", "Unknown", 2);
Console.WriteLine($"Contains(Unknown)? {transportSet.Contains(checkOther)}");    // false

// -------- Удаление --------
Console.WriteLine("\n=== Удаление ===\n");

transportSet.Remove(car2);
Console.WriteLine($"После удаления car2: {transportSet.Count} элементов.");

// -------- Объединение и пересечение --------
Console.WriteLine("\n=== Операции над множествами ===\n");

HashSet<Transport> setA = new HashSet<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Car("Camry", "Toyota", 4)
};

HashSet<Transport> setB = new HashSet<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Mountain Explorer", "Giant", true)
};

// Объединение
var union = new HashSet<Transport>(setA);
union.UnionWith(setB);
Console.WriteLine($"Union: {union.Count} элементов");

// Пересечение
var intersect = new HashSet<Transport>(setA);
intersect.IntersectWith(setB);
Console.WriteLine($"Intersect: {intersect.Count} элементов");

// Разность
var except = new HashSet<Transport>(setA);
except.ExceptWith(setB);
Console.WriteLine($"Except: {except.Count} элементов");

// -------- Подмножество --------
Console.WriteLine("\n=== Подмножество ===\n");

HashSet<Transport> small = new HashSet<Transport>
{
    new Car("Model S", "Tesla", 4)
};

Console.WriteLine($"small ⊆ setA? {small.IsSubsetOf(setA)}");   // true
Console.WriteLine($"setA ⊇ small? {setA.IsSupersetOf(small)}"); // true
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте при помощи  HashSet. 

````
using System;
using System.Collections.Generic;

public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }
    public string Color { get; set; }
    public bool IsWild { get; set; }
    public int Id { get; set; }

    public static int TotalAnimals { get; private set; } = 0;
    protected Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
        Habitat = "Неизвестно";
        Color = "Неизвестный";
        IsWild = true;
        Id = TotalAnimals;
    }

    protected Animal(string name, int age, double weight, string habitat) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    // -------- Equals / GetHashCode для HashSet --------
    // Два животных равны, если совпадают Name + Age + Habitat
    public override bool Equals(object obj)
    {
        if (obj is Animal other)
        {
            return Name == other.Name
                && Age == other.Age
                && Habitat == other.Habitat;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Age, Habitat);
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] #{Id} {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Среда: {Habitat}, Окрас: {Color}, Дикое: {IsWild}");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public int EnergyLevel { get; set; }
    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public int BarkVolume { get; set; }

    public Dog(string name, int age, double weight, string habitat, string breed)
        : base(name, age, weight, habitat)
    {
        Breed = breed;
        EnergyLevel = 100;
        IsTrained = false;
        FavoriteToy = "Мяч";
        BarkVolume = 70;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}, Дрессирован: {IsTrained}, Игрушка: {FavoriteToy}");
    }

    public void Train()
    {
        IsTrained = true;
        Console.WriteLine($"[Dog] {Name} теперь дрессирован!");
    }

    public void PlayWithToy()
    {
        EnergyLevel -= 8;
        Console.WriteLine($"[Dog] {Name} играет с «{FavoriteToy}». Энергия: {EnergyLevel}");
    }

    public void GuardHouse()
    {
        EnergyLevel -= 15;
        Console.WriteLine($"[Dog] {Name} охраняет дом. Энергия: {EnergyLevel}");
    }
}
public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }

    // 3 атрибута
    public bool IsNeutered { get; set; }
    public string FavoriteFood { get; set; }
    public int PurrVolume { get; set; }

    public Cat(string name, int age, double weight, string habitat, bool isIndoor)
        : base(name, age, weight, habitat)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        IsNeutered = false;
        FavoriteFood = "Рыба";
        PurrVolume = 50;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}, Стерилизован: {IsNeutered}, Еда: {FavoriteFood}");
    }

    public void EatFavorite()
    {
        Console.WriteLine($"[Cat] {Name} ест любимую еду «{FavoriteFood}».");
    }

    public void Scratch()
    {
        Console.WriteLine($"[Cat] {Name} точит когти.");
    }

    public void ClimbTree()
    {
        Console.WriteLine($"[Cat] {Name} залезла на дерево.");
    }
}
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    public int EggsPerYear { get; set; }
    public bool IsMigratory { get; set; }
    public string BeakType { get; set; }

    public Bird(string name, int age, double weight, string habitat,
                double wingspan, bool canFly)
        : base(name, age, weight, habitat)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        EggsPerYear = 5;
        IsMigratory = false;
        BeakType = "Короткий";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Размах: {Wingspan} м, Летает: {CanFly}, Яиц/год: {EggsPerYear}, Перелётная: {IsMigratory}");
    }

    public void LayEggs()
    {
        Console.WriteLine($"[Bird] {Name} откладывает яйца ({EggsPerYear} шт/год).");
    }

    public void MigrateSeason()
    {
        if (IsMigratory)
            Console.WriteLine($"[Bird] {Name} мигрирует на юг.");
        else
            Console.WriteLine($"[Bird] {Name} остаётся зимовать.");
    }

    public void Peck()
    {
        Console.WriteLine($"[Bird] {Name} клюёт корм.");
    }
}

Dog dog = new Dog("Рекс", 5, 25.0, "Дом", "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, "Дом", true);
Bird bird = new Bird("Кеша", 1, 0.4, "Лес", 0.3, true);

Dog dog2 = new Dog("Шарик", 3, 15.0, "Двор", "Овчарка");
Cat cat2 = new Cat("Барсик", 4, 5.5, "Дом", false);
Bird bird2 = new Bird("Попугай", 2, 0.35, "Лес", 0.25, true);

Console.WriteLine("HashSet<Animal>: добавление\n");

HashSet<Animal> animalSet = new HashSet<Animal>();

Console.WriteLine($"Добавить Рекса: {animalSet.Add(dog)}");       // true
Console.WriteLine($"Добавить Мурку: {animalSet.Add(cat)}");      // true
Console.WriteLine($"Добавить Кешу: {animalSet.Add(bird)}");      // true
Console.WriteLine($"Добавить Шарика: {animalSet.Add(dog2)}");    // true
Console.WriteLine($"Добавить Барсика: {animalSet.Add(cat2)}");   // true
Console.WriteLine($"Добавить Попугая: {animalSet.Add(bird2)}");  // true

Console.WriteLine($"\nВсего в HashSet: {animalSet.Count}");

Console.WriteLine("\nПопытка добавить дубликаты\n");

// Копия Рекса — те же Name + Age + Habitat
Dog duplicateDog = new Dog("Рекс", 5, 25.0, "Дом", "Лабрадор");
bool added1 = animalSet.Add(duplicateDog);
Console.WriteLine($"Дубликат Рекса добавлен? {added1}");   // false

// Копия Мурки
Cat duplicateCat = new Cat("Мурка", 2, 4.2, "Дом", true);
bool added2 = animalSet.Add(duplicateCat);
Console.WriteLine($"Дубликат Мурки добавлен? {added2}");   // false

// Другой Рекс — другой возраст → НЕ дубликат
Dog otherDog = new Dog("Рекс", 10, 30.0, "Дом", "Лабрадор");
bool added3 = animalSet.Add(otherDog);
Console.WriteLine($"Рекс 10 лет добавлен? {added3}");      // true

Console.WriteLine($"\nВсего в HashSet: {animalSet.Count}");
````

````
Console.WriteLine("\nПеребор HashSet\n");

foreach (Animal a in animalSet)
{
    a.DisplayInfo();
    a.MakeSound();
    Console.WriteLine();
}
````

````
Console.WriteLine("Проверка наличия (Contains)\n");

Dog checkDog = new Dog("Рекс", 5, 25.0, "Дом", "Лабрадор");   // такой же, как dog
Console.WriteLine($"Contains(Рекс 5 лет)? {animalSet.Contains(checkDog)}");   // true

Dog checkOther = new Dog("Тузик", 3, 10.0, "Дом", "Дворняга");
Console.WriteLine($"Contains(Тузик)? {animalSet.Contains(checkOther)}");      // false

Console.WriteLine("\nУдаление\n");
animalSet.Remove(dog2);
Console.WriteLine($"После удаления Шарика: {animalSet.Count} элементов.");

Console.WriteLine("\nОперации над множествами\n");

// Множество только собак
HashSet<Animal> dogSet = new HashSet<Animal>();
foreach (Animal a in animalSet)
    if (a is Dog)
        dogSet.Add(a);

Console.WriteLine($"Собак в dogSet: {dogSet.Count}");

// Множество только кошек
HashSet<Animal> catSet = new HashSet<Animal>();
foreach (Animal a in animalSet)
    if (a is Cat)
        catSet.Add(a);

Console.WriteLine($"Кошек в catSet: {catSet.Count}");

HashSet<Animal> union = new HashSet<Animal>(dogSet);
union.UnionWith(catSet);
Console.WriteLine($"Union (собаки + кошки): {union.Count}");

HashSet<Animal> intersect = new HashSet<Animal>(dogSet);
intersect.IntersectWith(catSet);
Console.WriteLine($"Intersect (собаки ∩ кошки): {intersect.Count}");   // 0

HashSet<Animal> except = new HashSet<Animal>(animalSet);
except.ExceptWith(dogSet);
Console.WriteLine($"Except (все − собаки): {except.Count}");

Console.WriteLine("\nПодмножество\n");

Console.WriteLine($"dogSet ⊆ animalSet? {dogSet.IsSubsetOf(animalSet)}");   // true
Console.WriteLine($"animalSet ⊇ dogSet? {animalSet.IsSupersetOf(dogSet)}"); // true
````

````
Console.WriteLine("\nОчистка\n");

HashSet<Animal> tempSet = new HashSet<Animal> { dog, cat };
Console.WriteLine($"tempSet.Count до очистки: {tempSet.Count}");
tempSet.Clear();
Console.WriteLine($"tempSet.Count после Clear(): {tempSet.Count}");
Console.WriteLine($"tempSet пуст? {tempSet.Count == 0}");

Console.WriteLine("\nСобственные методы\n");

dog.Train();
dog.PlayWithToy();
dog.GuardHouse();

cat.EatFavorite();
cat.Scratch();
cat.ClimbTree();

bird.LayEggs();
bird.MigrateSeason();
bird.Peck();

Console.WriteLine("\nСтатические члены\n");
Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals: {Animal.TotalAnimals}");
````
