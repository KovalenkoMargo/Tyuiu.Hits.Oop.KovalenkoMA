### Перекрытие методов в C#

Перекрытие методов (или переопределение) — это процесс, при котором производный класс предоставляет новую реализацию метода, уже существующего в его базовом классе. Это позволяет изменить или расширить поведение метода в производном классе без изменения исходного метода в базовом классе. Перекрытие методов является ключевой особенностью полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, где каждый из этих классов имеет метод `Drive()`, который принимает параметр `speed`.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive()
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

В этом примере, метод `Drive()` в классе `Car` и `Motorcycle` переопределяет метод `Drive()` из базового класса `Vehicle`. Это означает, что при вызове метода `Drive()` на объекте типа `Car` или `Motorcycle`, будет вызвана соответствующая реализация метода в этих классах, а не в классе `Vehicle`.

#### Отличие перекрытия от перегрузки методов

Перегрузка методов — это процесс определения нескольких методов с одним и тем же именем, но с разным количеством или типом параметров. Например:

```csharp
public class ExampleClass
{
    public void Print(int number)
    {
        Console.WriteLine(number);
    }

    public void Print(double number)
    {
        Console.WriteLine(number.ToString());
    }
}
```

В этом случае, у нас есть два метода с именем `Print`, но с разным типом параметра. Это отличается от перекрытия методов, где метод в производном классе полностью заменяет метод в базовом классе.

#### Вызов методов родителя

Иногда в производном классе может потребоваться вызвать метод родителя. Для этого используется ключевое слово `base`. Например, если мы хотим добавить дополнительную логику в метод `Drive()` класса `Car`, сохраняя при этом оригинальное поведение метода `Drive()` из класса `Vehicle`.

```csharp
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed); // Вызов метода Drive() из класса Vehicle
        Console.WriteLine("Добавляем специфическое поведение автомобиля");
    }
}
```

В этом примере, после вызова `base.Drive(speed)`, выполняется дополнительная логика, специфичная для класса `Car`.


### Перегрузка методов в C#

Перегрузка методов — это способность класса определить несколько методов с одним и тем же именем, но с разным количеством или типом параметров. Это позволяет разработчикам предоставлять различные варианты выполнения одного и того же действия в зависимости от входных данных. Перегрузка методов является одной из форм выражения полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример, где класс `Vehicle` содержит перегруженные методы `Drive()`, которые принимают разные типы параметров.

```csharp
public class Vehicle
{
    // Метод Drive(), принимающий int
    public void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }

    // Метод Drive(), принимающий string
    public void Drive(string mode)
    {
        Console.WriteLine($"Автомобиль движется в {mode} режиме.");
    }

    // Метод Drive(), принимающий bool
    public void Drive(bool isManual)
    {
        if (isManual)
        {
            Console.WriteLine("Автомобиль переключает передачи вручную.");
        }
        else
        {
            Console.WriteLine("Автомобиль оснащен автоматической коробкой передач.");
        }
    }
}
```

В этом примере класс `Vehicle` определяет три метода с именем `Drive()`, каждый из которых принимает разные типы параметров: `int`, `string` и `bool`. Это позволяет вызывать разные версии метода `Drive()` в зависимости от контекста, например:

```csharp
Vehicle myVehicle = new Vehicle();

myVehicle.Drive(60); // Вызов метода Drive() с int параметром
myVehicle.Drive("Eco"); // Вызов метода Drive() с string параметром
myVehicle.Drive(true); // Вызов метода Drive() с bool параметром
```

#### Отличие перегрузки от переопределения

Отличие перегрузки методов от переопределения заключается в том, что перегрузка методов включает определение нескольких методов с одним именем, но с разным набором параметров, тогда как переопределение методов (или перекрытие) означает предоставление новой реализации метода в производном классе, который уже существует в базовом классе.

#### Вызов методов

При вызове перегруженного метода компилятор выбирает подходящую версию метода на основе типов и количества аргументов, переданных в вызове. Если типы и количество аргументов совпадают с несколькими версиями метода, компилятор выберет ту версию, которая была объявлена первой.

### Заключение

Перегрузка методов в C# позволяет классам предоставлять различные варианты выполнения одного и того же действия, адаптируясь к различным условиям использования. Это мощный инструмент для создания гибких и удобных API, позволяющих клиентам класса использовать его функциональность наиболее эффективно.

Перекрытие методов в C# позволяет производным классам изменять или расширять поведение методов базовых классов, что является ключевой особенностью полиморфизма. Это отличается от перегрузки методов, которая предполагает определение нескольких методов с одинаковым именем, но с разным набором параметров.



<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке:</h4>

----

````
using System;

public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}

public class CarWithBase : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed);                       // вызов версии Vehicle
        Console.WriteLine("Добавляем специфическое поведение автомобиля");
    }
}

public class VehicleOverloaded
{
    // Drive(int)
    public void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }

    // Drive(string)
    public void Drive(string mode)
    {
        Console.WriteLine($"Автомобиль движется в {mode} режиме.");
    }

    // Drive(bool)
    public void Drive(bool isManual)
    {
        if (isManual)
            Console.WriteLine("Автомобиль переключает передачи вручную.");
        else
            Console.WriteLine("Автомобиль оснащен автоматической коробкой передач.");
    }
}

public class ExampleClass
{
    public void Print(int number)
    {
        Console.WriteLine($"int: {number}");
    }

    public void Print(double number)
    {
        Console.WriteLine($"double: {number}");
    }
}


Console.WriteLine("=== A. Перекрытие методов (override) ===\n");

Vehicle v = new Vehicle();
Car car = new Car();
Motorcycle moto = new Motorcycle();

v.Drive(60);
car.Drive(80);
moto.Drive(100);

Console.WriteLine("\n=== Полиморфизм через Vehicle[] ===\n");

Vehicle[] vehicles = { v, car, moto };
foreach (Vehicle item in vehicles)
{
    item.Drive(90);
}

Console.WriteLine("\n=== Вызов base.Drive() ===\n");

CarWithBase carBase = new CarWithBase();
carBase.Drive(70);

Console.WriteLine("\n=== B. Перегрузка методов (overload) ===\n");

VehicleOverloaded ov = new VehicleOverloaded();
ov.Drive(60);          // int
ov.Drive("Eco");       // string
ov.Drive(true);        // bool
ov.Drive(false);       // bool

Console.WriteLine("\n=== C. Перегрузка Print(int) / Print(double) ===\n");

ExampleClass ex = new ExampleClass();
ex.Print(42);          // int
ex.Print(3.14);        // double
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также используйте перекрытия и перегрузки методов.

````
using System;

public class Animal
{
    public string Name
    { get; set;}

    public int Age
    { get; set;}
    public double Weight
    { get; set;}

    public static int TotalAnimals { get; private set; } = 0;

    public Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
        Console.WriteLine($"[Animal] Создано по умолчанию. Всего: {TotalAnimals}");
    }

    public Animal(string name, int age, double weight) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Console.WriteLine($"[Animal] Создано: {Name}, {Age} лет, {Weight} кг");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг");
    }

    public void Eat(double amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество еды.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    // Eat(string) — покормить конкретным кормом
    public void Eat(string food)
    {
        if (string.IsNullOrWhiteSpace(food))
        {
            Console.WriteLine("[Animal] Некорректный корм.");
            return;
        }
        Weight += 0.2;
        Console.WriteLine($"[Animal] {Name} поел «{food}». Новый вес: {Weight:F2} кг");
    }

    // Eat(double, string) — покормить кормом с количеством
    public void Eat(double amount, string food)
    {
        if (amount <= 0 || string.IsNullOrWhiteSpace(food))
        {
            Console.WriteLine("[Animal] Некорректные параметры кормления.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел {amount} кг «{food}». Новый вес: {Weight:F2} кг");
    }

    public void Move(int distance)
    {
        Console.WriteLine($"[Animal] {Name} перемещается на {distance} м.");
    }

    // Move(int, string) — пройти distance метров заданным способом
    public void Move(int distance, string mode)
    {
        Console.WriteLine($"[Animal] {Name} перемещается на {distance} м способом «{mode}».");
    }

    // Move(string) — переместиться в направлении
    public void Move(string direction)
    {
        Console.WriteLine($"[Animal] {Name} перемещается в направлении «{direction}».");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal
{
    public string Breed
    { get; set;}

    public int EnergyLevel
    { get; set;}

    public Dog() : base()
    {
        Breed = "Дворняга";
        EnergyLevel = 100;
        Console.WriteLine("[Dog] Создана собака по умолчанию.");
    }

    public Dog(string name, int age, double weight, string breed)
        : base(name, age, weight)
    {
        Breed = breed;
        EnergyLevel = 100;
        Console.WriteLine($"[Dog] Создана собака: {Name}, порода: {Breed}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}");
    }


    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }

    public void BringSlippers(string item)
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} принёс «{item}»! Энергия: {EnergyLevel}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor
   { get; set;}

    public int LivesLeft
    { get; set;}

    public Cat() : base()
    {
        IsIndoor = true;
        LivesLeft = 9;
        Console.WriteLine("[Cat] Создан кот по умолчанию.");
    }

    public Cat(string name, int age, double weight, bool isIndoor)
        : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Console.WriteLine($"[Cat] Создан кот: {Name}, домашний: {IsIndoor}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}");
    }

    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }

    public void Purr(int minutes)
    {
        Console.WriteLine($"[Cat] {Name} мурлычет {minutes} минут...");
    }
}

public class Bird : Animal
{
    public double Wingspan
    { get; set;}

    public bool CanFly
    { get; set;}

    public Bird() : base()
    {
        Wingspan = 0.3;
        CanFly = true;
        Console.WriteLine("[Bird] Создана птица по умолчанию.");
    }

    public Bird(string name, int age, double weight, double wingspan, bool canFly)
        : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        Console.WriteLine($"[Bird] Создана птица: {Name}, размах: {Wingspan} м");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} поёт: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Размах: {Wingspan} м, Умеет летать: {CanFly}");
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает!");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }

    public void Fly(int height)
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает на высоте {height} м.");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать, но мечтает о {height} м.");
    }
}

Animal genericAnimal = new Animal("Зверь", 3, 10.5);
Dog dog = new Dog("Рекс", 5, 25.0, "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, true);
Bird bird = new Bird("Кеша", 1, 0.4, 0.3, true);
````

````
Console.WriteLine("\n=== ПЕРЕКРЫТИЕ (override): DisplayInfo ===\n");

genericAnimal.DisplayInfo();  Console.WriteLine();
dog.DisplayInfo();            Console.WriteLine();
cat.DisplayInfo();            Console.WriteLine();
bird.DisplayInfo();           Console.WriteLine();

Console.WriteLine("=== ПЕРЕКРЫТИЕ (override): MakeSound ===\n");

Animal[] animals = { genericAnimal, dog, cat, bird };
foreach (Animal a in animals)
{
    a.MakeSound();
}
````

````
Console.WriteLine("\n=== ПЕРЕГРУЗКА (overload) базового класса: Eat ===\n");

genericAnimal.Eat(5.0);                // Eat(double)
genericAnimal.Eat("мясо");             // Eat(string)
genericAnimal.Eat(3.0, "рыба");        // Eat(double, string)

Console.WriteLine("\n=== ПЕРЕГРУЗКА (overload) базового класса: Move ===\n");

genericAnimal.Move(100);               // Move(int)
genericAnimal.Move(50, "бег");         // Move(int, string)
genericAnimal.Move("север");           // Move(string)

Console.WriteLine("\n=== ПЕРЕГРУЗКА (overload) в Dog: BringSlippers ===\n");

dog.BringSlippers();                   // без параметров
dog.BringSlippers("газету");           // со строкой

````

````
Console.WriteLine("\n=== ПЕРЕГРУЗКА (overload) в Cat: Purr ===\n");

cat.Purr();                            // без параметров
cat.Purr(5);                           // с числом

Console.WriteLine("\n=== ПЕРЕГРУЗКА (overload) в Bird: Fly ===\n");

bird.Fly();                            // без параметров
bird.Fly(200);                         // с высотой

Console.WriteLine("\n=== ПЕРЕКРЫТИЕ + base: DisplayInfo производных ===\n");
dog.DisplayInfo();
Console.WriteLine();
cat.DisplayInfo();
Console.WriteLine();
bird.DisplayInfo();

````

````
Console.WriteLine("\n=== Взаимодействие через базовый тип ===\n");

foreach (Animal a in animals)
{
    a.DisplayInfo();
    a.MakeSound();
    Console.WriteLine();
}

Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals (через свойство): {Animal.TotalAnimals}");
````
