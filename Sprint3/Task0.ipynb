### Понятие полиморфизма в C# с параметрами

Полиморфизм в объектно-ориентированном программировании позволяет объектам разных классов обрабатывать одни и те же сообщения по-разному. В C#, полиморфизм часто реализуется через наследование и переопределение методов. Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, чтобы продемонстрировать, как полиморфизм работает с параметрами.

#### Определение классов

Сначала определим базовый класс `Vehicle` с виртуальным методом `Drive()`, который принимает параметры. Затем создадим два производных класса: `Car` и `Motorcycle`, которые переопределят этот метод.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive(), принимающим параметры
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

#### Использование полиморфизма

Теперь создадим список объектов типа `Vehicle` и добавим в него экземпляры `Car` и `Motorcycle`. Затем вызовем метод `Drive()` для каждого элемента списка, передавая разные скорости. Это демонстрирует, как полиморфизм позволяет нам работать с объектами разных типов через общий интерфейс, но при этом выполнять различные действия в зависимости от "реального" типа объекта.

```csharp
List<Vehicle> vehicles = new List<Vehicle>
{
    new Car(),
    new Motorcycle(),
    new Vehicle() // Добавляем экземпляр базового класса для демонстрации
};

foreach (var vehicle in vehicles)
{
    vehicle.Drive(100); // Вызываем метод Drive() 
}

```

В результате выполнения этого кода, каждый объект в списке вызывает свою версию метода `Drive()`, демонстрируя, что при работе с объектами через базовый класс (`Vehicle`) можно достичь различных действий в зависимости от "реального" типа объекта, даже когда используются разные скорости.

Этот пример иллюстрирует ключевое понятие полиморфизма в ООП: возможность работы с объектами разных типов через общий интерфейс или базовый класс, что позволяет писать более гибкий и расширяемый код.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке:</h4>

----

````
using System;
using System.Collections.Generic;

public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}


Console.WriteLine("=== Полиморфизм через базовый тип Vehicle ===\n");

List<Vehicle> vehicles = new List<Vehicle>
{
    new Car(),
    new Motorcycle(),
    new Vehicle()   
};

foreach (var vehicle in vehicles)
{
    vehicle.Drive(100);   // у каждого своя версия Drive()
}

Console.WriteLine("\n=== Разные скорости через общий интерфейс ===\n");

int[] speeds = { 60, 90, 120 };
foreach (var v in vehicles)
{
    foreach (int s in speeds)
    {
        v.Drive(s);
    }
    Console.WriteLine();
}

Console.WriteLine("=== Полиморфизм в действии ===\n");

Vehicle v1 = new Car();
Vehicle v2 = new Motorcycle();
Vehicle v3 = new Vehicle();

v1.Drive(80);   // Автомобиль
v2.Drive(80);   // Мотоцикл
v3.Drive(80);   // Транспорт
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также механизм полиморфизм.

````
using System;
using System.Collections.Generic;

public class Animal
{
    public string Name
    { get; set;}

    public int Age
    { get; set;}

    public double Weight
    { get; set;}

    public static int TotalAnimals { get; private set; } = 0;

    public Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
        Console.WriteLine($"[Animal] Создано по умолчанию. Всего: {TotalAnimals}");
    }

    public Animal(string name, int age, double weight) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Console.WriteLine($"[Animal] Создано: {Name}, {Age} лет, {Weight} кг");
    }
    public virtual void MakeSound(int volume) // volume: громкость звука (0–100). > 70 → громкий вариант, иначе → тихий.
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук громкостью {volume}.");
    }

    public virtual void Move(int distance)// distance: дистанция в метрах. Dog тратит энергию, Cat идёт грациозно, Bird летит (если CanFly).
    {
        Console.WriteLine($"[Animal] {Name} перемещается на {distance} м.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг");
    }

    public void Eat(double amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество еды.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}


public class Dog : Animal
{
    public string Breed
    { get; set;}

    public int EnergyLevel
    { get; set;}

    public Dog() : base()
    {
        Breed = "Дворняга";
        EnergyLevel = 100;
        Console.WriteLine("[Dog] Создана собака по умолчанию.");
    }

    public Dog(string name, int age, double weight, string breed)
        : base(name, age, weight)
    {
        Breed = breed;
        EnergyLevel = 100;
        Console.WriteLine($"[Dog] Создана собака: {Name}, порода: {Breed}");
    }

    public override void MakeSound(int volume)
    {
        string sound = volume > 70 ? "ГАВ-ГАВ-ГАВ!" : "гав-гав";
        Console.WriteLine($"[Dog] {Name} лает (громкость {volume}): {sound}");
    }

    public override void Move(int distance)
    {
        EnergyLevel -= distance / 2;
        Console.WriteLine($"[Dog] {Name} бежит {distance} м на четырёх лапах. Энергия: {EnergyLevel}");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}");
    }

    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }
}

public class Cat : Animal
{

    public bool IsIndoor
    { get; set;}

    public int LivesLeft
    { get; set;}

    public Cat() : base()
    {
        IsIndoor = true;
        LivesLeft = 9;
        Console.WriteLine("[Cat] Создан кот по умолчанию.");
    }

    public Cat(string name, int age, double weight, bool isIndoor)
        : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Console.WriteLine($"[Cat] Создан кот: {Name}, домашний: {IsIndoor}");
    }

    public override void MakeSound(int volume)
    {
        string sound = volume > 70 ? "МЯУ-МЯУ!" : "мяу";
        Console.WriteLine($"[Cat] {Name} мяукает (громкость {volume}): {sound}");
    }

    public override void Move(int distance)
    {
        Console.WriteLine($"[Cat] {Name} грациозно идёт {distance} м. Жизней: {LivesLeft}");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}");
    }

    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }
}

public class Bird : Animal
{
    public double Wingspan
    { get; set;}

    public bool CanFly
    { get; set;}

    public Bird() : base()
    {
        Wingspan = 0.3;
        CanFly = true;
        Console.WriteLine("[Bird] Создана птица по умолчанию.");
    }

    public Bird(string name, int age, double weight, double wingspan, bool canFly)
        : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        Console.WriteLine($"[Bird] Создана птица: {Name}, размах: {Wingspan} м");
    }

    public override void MakeSound(int volume)
    {
        string sound = volume > 70 ? "ЧИК-ЧИРИК-ЧИК!" : "чик-чирик";
        Console.WriteLine($"[Bird] {Name} поёт (громкость {volume}): {sound}");
    }

    public override void Move(int distance)
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} пролетела {distance} м (размах {Wingspan} м).");
        else
            Console.WriteLine($"[Bird] {Name} прошла {distance} м пешком.");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Размах: {Wingspan} м, Умеет летать: {CanFly}");
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает!");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }
}

Animal genericAnimal = new Animal("Зверь", 3, 10.5);
Dog dog = new Dog("Рекс", 5, 25.0, "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, true);
Bird bird = new Bird("Кеша", 1, 0.4, 0.3, true);
````

````
Console.WriteLine("\nполиморфизм - массив Animal\n");

Animal[] animals = { genericAnimal, dog, cat, bird};

// Вызов через базовый тип — у каждого своя версия
foreach (Animal a in animals)
{
    a.MakeSound(50);   // тихо
    a.MakeSound(90);   // громко
    a.Move(100);
    Console.WriteLine();
}
````

````
Console.WriteLine("полиморфизм через List<Animal>\n");

List<Animal> animalList = new List<Animal>
{
    new Dog("Шарик", 3, 15.0, "Овчарка"),
    new Cat("Барсик", 4, 5.5, false),
    new Bird("Попугай", 2, 0.35, 0.25, true)
};

foreach (var a in animalList)
{
    a.MakeSound(80);
    a.Move(50);
    a.DisplayInfo();
    Console.WriteLine();
}
````

````
Console.WriteLine("полимор с разными параметрами\n");

int[] volumes = { 30, 60, 100 };
foreach (Animal a in animals)
{
    Console.WriteLine($"--- {a.GetType().Name}: {a.Name} ---");
    foreach (int v in volumes)
    {
        a.MakeSound(v);
    }
    Console.WriteLine();
}
````
