### Структура и Объявление Класса в C#

В C#, класс объявляется с использованием ключевого слова `class`, за которым следует имя класса. Класс может содержать поля (данные), свойства (способ доступа к данным), методы (функциональность) и события. 

#### Свойства

Свойства в C# используются для инкапсуляции данных. Они предоставляют гибкий механизм для чтения, записи или вычисления значений приватных полей. Свойства могут иметь два акцессора: `get` (геттер) для чтения значения и `set` (сеттер) для его изменения.

#### Геттеры и Сеттеры

Геттеры и сеттеры являются частью свойств класса в C#. Геттер (`get`) используется для возврата значения свойства, а сеттер (`set`) — для установки нового значения. Сеттер может включать логику валидации или другие операции перед изменением значения поля.

### Пример Классов Vehicle и Car

Давайте рассмотрим примеры классов `Vehicle` и `Car`, чтобы проиллюстрировать использование свойств, геттеров и сеттеров, а также разницу между статическими и обычными атрибутами и методами.

#### Класс Vehicle

```csharp
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}
```

#### Класс Car

```csharp
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}
```

### Разница между Статическими и Обычными Атрибутами и Методами Класса

- **Статические атрибуты и методы** принадлежат самому классу, а не его экземплярам. Они могут быть вызваны без создания объекта класса. В примере выше `TotalVehicles` является статическим атрибутом, а `DisplayTotalVehicles()` — статическим методом класса `Vehicle`.
- **Обычные атрибуты и методы** принадлежат экземплярам класса и могут иметь разные значения для разных объектов. В примере выше `_color`, `_speed` и `_fuelLevel` являются обычными атрибутами, а методы `Accelerate()` и `Refuel()` — обычными методами.

````
using System;

// ==================== Класс Vehicle ====================
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером (с валидацией)
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}

// ==================== Класс Car ====================
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}

// ==================== Демонстрация (top-level код) ====================

Console.WriteLine("=== Создание объектов ===\n");

Vehicle v1 = new Vehicle();
Vehicle v2 = new Vehicle();
Car car1 = new Car();

Console.WriteLine("\n=== Работа со свойствами ===\n");

v1.Color = "Красный";
v1.Speed = 60;
Console.WriteLine($"v1: Color = {v1.Color}, Speed = {v1.Speed}");

car1.Color = "Синий";
car1.Speed = 40;
car1.Accelerate(20);
Console.WriteLine($"car1: Color = {car1.Color}, Speed = {car1.Speed}, FuelLevel = {car1.FuelLevel}");

car1.Refuel(30);
Console.WriteLine($"car1 после заправки: FuelLevel = {car1.FuelLevel}");

Console.WriteLine("\n=== Проверка валидации сеттера ===\n");
try
{
    v1.Speed = -10;   // должно бросить исключение
}
catch (ArgumentOutOfRangeException ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}

Console.WriteLine("\n=== Статические члены ===\n");
Vehicle.DisplayTotalVehicles();
Console.WriteLine($"TotalVehicles (через свойство): {Vehicle.TotalVehicles}");
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса

````
using System;

// ==================== Базовый класс Animal ====================
public class Animal
{
    // -------- Приватные поля (инкапсуляция) --------
    private string _name;
    private int _age;
    private double _weight;

    // -------- Свойство Name с геттером и сеттером --------
    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым!");
            _name = value;
        }
    }

    // -------- Свойство Age с геттером и сеттером --------
    public int Age
    {
        get { return _age; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
            _age = value;
        }
    }

    // -------- Свойство Weight с геттером и сеттером --------
    public double Weight
    {
        get { return _weight; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("Вес должен быть положительным!");
            _weight = value;
        }
    }

    // -------- Статический атрибут --------
    public static int TotalAnimals { get; private set; } = 0;

    // ==================== Конструкторы ====================
    public Animal()
    {
        TotalAnimals++;
        _name = "Безымянное";
        _age = 0;
        _weight = 1.0;
        Console.WriteLine($"[Animal] Создано по умолчанию. Всего: {TotalAnimals}");
    }

    public Animal(string name, int age, double weight) : this()
    {
        Name = name;      // через сеттер — с валидацией
        Age = age;
        Weight = weight;
        Console.WriteLine($"[Animal] Создано: {Name}, {Age} лет, {Weight} кг");
    }

    // ==================== Обычные методы ====================
    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public void Eat(double amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество еды.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    public void ShowInfo()
    {
        Console.WriteLine($"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2}");
    }

    // ==================== Статический метод ====================
    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}

// ==================== Производный класс Dog ====================
public class Dog : Animal
{
    private string _breed;
    private int _energyLevel;

    // Свойство Breed с геттером и сеттером
    public string Breed
    {
        get { return _breed; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Порода не может быть пустой!");
            _breed = value;
        }
    }

    // Свойство EnergyLevel с геттером и сеттером
    public int EnergyLevel
    {
        get { return _energyLevel; }
        set
        {
            if (value < 0) _energyLevel = 0;
            else if (value > 100) _energyLevel = 100;
            else _energyLevel = value;
        }
    }

    // Конструкторы
    public Dog() : base()
    {
        _breed = "Дворняга";
        _energyLevel = 100;
        Console.WriteLine("[Dog] Создана собака по умолчанию.");
    }

    public Dog(string name, int age, double weight, string breed)
        : base(name, age, weight)
    {
        Breed = breed;
        EnergyLevel = 100;
        Console.WriteLine($"[Dog] Создана собака: {Name}, порода: {Breed}");
    }

    // Переопределение метода (полиморфизм)
    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    // Обычный метод
    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }

    // Обычный метод
    public void ShowDogInfo()
    {
        ShowInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}");
    }
}

// ==================== Производный класс Cat ====================
public class Cat : Animal
{
    private bool _isIndoor;
    private int _livesLeft;

    // Свойство IsIndoor с геттером и сеттером
    public bool IsIndoor
    {
        get { return _isIndoor; }
        set { _isIndoor = value; }
    }

    // Свойство LivesLeft с геттером и сеттером (с валидацией)
    public int LivesLeft
    {
        get { return _livesLeft; }
        set
        {
            if (value < 0 || value > 9)
                throw new ArgumentOutOfRangeException("Жизней у кота от 0 до 9!");
            _livesLeft = value;
        }
    }

    // Конструкторы
    public Cat() : base()
    {
        _isIndoor = true;
        _livesLeft = 9;
        Console.WriteLine("[Cat] Создан кот по умолчанию.");
    }

    public Cat(string name, int age, double weight, bool isIndoor)
        : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Console.WriteLine($"[Cat] Создан кот: {Name}, домашний: {IsIndoor}");
    }

    // Переопределение метода (полиморфизм)
    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    // Обычный метод
    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }

    // Обычный метод
    public void ShowCatInfo()
    {
        ShowInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}");
    }
}

// ==================== Производный класс Parrot ====================
public class Parrot : Animal
{
    private string _phrase;
    private bool _canFly;

    // Свойство Phrase с геттером и сеттером
    public string Phrase
    {
        get { return _phrase; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Фраза не может быть пустой!");
            _phrase = value;
        }
    }

    // Свойство CanFly с геттером и сеттером
    public bool CanFly
    {
        get { return _canFly; }
        set { _canFly = value; }
    }

    // Конструкторы
    public Parrot() : base()
    {
        _phrase = "Кеша хороший";
        _canFly = true;
        Console.WriteLine("[Parrot] Создан попугай по умолчанию.");
    }

    public Parrot(string name, int age, double weight, string phrase)
        : base(name, age, weight)
    {
        Phrase = phrase;
        CanFly = true;
        Console.WriteLine($"[Parrot] Создан попугай: {Name}, фраза: {Phrase}");
    }

    // Переопределение метода (полиморфизм)
    public override void MakeSound()
    {
        Console.WriteLine($"[Parrot] {Name} говорит: {Phrase}!");
    }

    // Обычный метод
    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Parrot] {Name} летает!");
        else
            Console.WriteLine($"[Parrot] {Name} не умеет летать.");
    }

    // Обычный метод
    public void ShowParrotInfo()
    {
        ShowInfo();
        Console.WriteLine($"[Parrot] Фраза: {Phrase}, Умеет летать: {CanFly}");
    }
}

// ==================== Демонстрация (top-level код) ====================

Console.WriteLine("=== Создание животных ===\n");

Animal genericAnimal = new Animal("Зверь", 3, 10.5);
Dog dog = new Dog("Рекс", 5, 25.0, "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, true);
Parrot parrot = new Parrot("Кеша", 1, 0.4, "Кеша хороший");

Console.WriteLine("\n=== Полиморфизм: MakeSound ===\n");
genericAnimal.MakeSound();
dog.MakeSound();
cat.MakeSound();
parrot.MakeSound();

Console.WriteLine("\n=== Обычные методы производных классов ===\n");
dog.BringSlippers();
cat.Purr();
parrot.Fly();

Console.WriteLine("\n=== Свойства и состояние ===\n");
dog.ShowDogInfo();
cat.ShowCatInfo();
parrot.ShowParrotInfo();

Console.WriteLine("\n=== Изменение свойств через сеттеры ===\n");
dog.EnergyLevel = 50;
cat.LivesLeft = 7;
parrot.Phrase = "Попка дурак";

Console.WriteLine($"dog.EnergyLevel = {dog.EnergyLevel}");
Console.WriteLine($"cat.LivesLeft = {cat.LivesLeft}");
Console.WriteLine($"parrot.Phrase = {parrot.Phrase}");

Console.WriteLine("\n=== Проверка валидации сеттеров ===\n");
try { cat.LivesLeft = 15; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { dog.Name = ""; }
catch (ArgumentException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { genericAnimal.Age = -5; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

Console.WriteLine("\n=== Статические члены ===\n");
Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals (через свойство): {Animal.TotalAnimals}");
````
