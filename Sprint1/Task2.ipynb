<h1 style="color:DodgerBlue">Создание экземпляров классов (объектов) и взаимодействие объектов между собой</h1>

### Создание экземпляров классов (объектов)

В C#, экземпляр класса создается с помощью оператора `new`, который выделяет память для нового объекта и вызывает конструктор класса для инициализации этого объекта. Конструктор — это специальный метод в классе, который имеет то же имя, что и класс, и используется для установки начального состояния объекта.

#### Пример создания экземпляров классов Vehicle и Car

```csharp
Vehicle myVehicle = new Vehicle();
Car myCar = new Car();
```

В этом примере создаются два объекта: один для класса `Vehicle` и другой для класса `Car`. Конструкторы этих классов будут автоматически вызваны при создании объектов.

### Взаимодействие объектов между Собой

Взаимодействие объектов между собой означает, что один объект может вызывать методы другого объекта или обращаться к его свойствам (если они доступны). Это позволяет объектам обмениваться информацией и координировать свои действия.

#### Пример взаимодействия объектов

```csharp
myVehicle.Color = "Red";
myVehicle.Speed = 60;

myCar.Color = "Blue";
myCar.FuelLevel = 100;
myCar.Refuel(50); // Увеличиваем уровень топлива на 50
```

В этом примере мы устанавливаем свойства `Color` и `Speed` для объекта `myVehicle`, а также свойства `Color` и `FuelLevel` для объекта `myCar`. Затем мы вызываем метод `Refuel` у объекта `myCar`, чтобы увеличить его уровень топлива.

### Закрепление правил использования модификаторов доступа

Модификаторы доступа определяют уровень доступности членов класса из других частей кода. Важно правильно использовать модификаторы доступа для обеспечения безопасности и инкапсуляции данных.

#### Примеры модификаторов доступа в классах Vehicle и Car

```csharp
public class Vehicle
{
    public string Color { get; set; } // public свойство
    private int speed; // private поле

    protected void SetSpeed(int value) // protected метод
    {
        speed = value;
    }
}

public class Car : Vehicle
{
    private int fuelLevel; // private поле

    public void Refuel(int amount) // public метод
    {
        fuelLevel += amount;
    }

    protected void CheckSpeedLimit() // protected метод
    {
        if (speed > 120)
        {
            Console.WriteLine("Превышен скоростной режим!");
        }
    }
}
```

В этом примере:
- Свойство `Color` в классе `Vehicle` объявлено как `public`, что позволяет доступ к нему из любого места в коде.
- Поле `speed` в классе `Vehicle` объявлено как `private`, что ограничивает доступ к нему только внутри класса `Vehicle`.
- Метод `SetSpeed` в классе `Vehicle` объявлен как `protected`, что позволяет доступ к нему из класса `Vehicle` и его производных классов.
- Поле `fuelLevel` в классе `Car` объявлено как `private`, ограничивая доступ к нему только внутри класса `Car`.
- Методы `Refuel` и `CheckSpeedLimit` в классе `Car` демонстрируют использование модификаторов доступа для контроля над тем, как объекты могут взаимодействовать друг с другом.

### Заключение

Создание экземпляров классов и взаимодействие между объектами являются ключевыми аспектами объектно-ориентированного программирования в C#. Правильное использование модификаторов доступа обеспечивает безопасность данных и инкапсуляцию, позволяя разработчикам создавать более надежные и управляемые программы.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Car и Vehicle ниже в блоке:</h4>

----

````
using System;

// ==================== Класс Vehicle ====================
public class Vehicle
{
    private int speed;                     // private поле

    public string Color { get; set; }      // public свойство

    // protected метод — доступен в Vehicle и наследниках
    protected void SetSpeed(int value)
    {
        speed = value;
        Console.WriteLine($"[Vehicle] Скорость установлена: {speed}");
    }

    // public метод для внешнего управления скоростью через protected-метод
    public void Accelerate(int delta)
    {
        SetSpeed(speed + delta);
    }

    // public свойство для чтения скорости (speed остаётся private)
    public int Speed
    {
        get { return speed; }
    }
}

// ==================== Класс Car ====================
public class Car : Vehicle
{
    private int fuelLevel;                 // private поле

    public int FuelLevel
    {
        get { return fuelLevel; }
        set { fuelLevel = value; }
    }

    // public метод
    public void Refuel(int amount)
    {
        fuelLevel += amount;
        Console.WriteLine($"[Car] Топливо: {fuelLevel}");
    }

    // protected метод — доступен в Car и наследниках Car
    protected void CheckSpeedLimit()
    {
        if (Speed > 120)
        {
            Console.WriteLine("[Car] Превышен скоростной режим!");
        }
        else
        {
            Console.WriteLine($"[Car] Скорость в норме: {Speed}");
        }
    }

    // public метод-обёртка, чтобы продемонстрировать protected CheckSpeedLimit
    public void CheckLimit()
    {
        CheckSpeedLimit();
    }
}

// ==================== Демонстрация (top-level код) ====================

Console.WriteLine("=== Создание экземпляров классов ===\n");

Vehicle myVehicle = new Vehicle();
Car myCar = new Car();

Console.WriteLine("\n=== Взаимодействие объектов ===\n");

myVehicle.Color = "Red";
myVehicle.Accelerate(60);

myCar.Color = "Blue";
myCar.FuelLevel = 100;
myCar.Refuel(50);      // Увеличиваем уровень топлива на 50

Console.WriteLine($"\nmyVehicle: Color = {myVehicle.Color}, Speed = {myVehicle.Speed}");
Console.WriteLine($"myCar: Color = {myCar.Color}, FuelLevel = {myCar.FuelLevel}");

Console.WriteLine("\n=== Проверка protected-метода CheckSpeedLimit ===\n");
myCar.Accelerate(70);   // теперь 70
myCar.CheckLimit();     // в норме

myCar.Accelerate(60);   // теперь 130
myCar.CheckLimit();     // превышение
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса с реализацией взаимодействия между объектами.

````
using System;

// ==================== Базовый класс Animal ====================
public class Animal
{
    // -------- Приватные поля --------
    private string _name;
    private int _age;
    private double _weight;

    // -------- Свойство Name (get/set с валидацией) --------
    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым!");
            _name = value;
        }
    }

    // -------- Свойство Age (get/set с валидацией) --------
    public int Age
    {
        get { return _age; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
            _age = value;
        }
    }

    // -------- Свойство Weight (get/set с валидацией) --------
    public double Weight
    {
        get { return _weight; }
        set
        {
            if (value <= 0)
                throw new ArgumentOutOfRangeException("Вес должен быть положительным!");
            _weight = value;
        }
    }

    // -------- Статический атрибут --------
    public static int TotalAnimals { get; private set; } = 0;

    // ==================== Конструкторы ====================
    public Animal()
    {
        TotalAnimals++;
        _name = "Безымянное";
        _age = 0;
        _weight = 1.0;
        Console.WriteLine($"[Animal] Создано по умолчанию. Всего: {TotalAnimals}");
    }

    public Animal(string name, int age, double weight) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Console.WriteLine($"[Animal] Создано: {Name}, {Age} лет, {Weight} кг");
    }

    // ==================== Обычные методы ====================
    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public void Eat(double amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("[Animal] Некорректное количество еды.");
            return;
        }
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    public void ShowInfo()
    {
        Console.WriteLine($"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2}");
    }

    // ==================== Статический метод ====================
    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }

    // ==================== Взаимодействие: метод, принимающий другого Animal ====================
    public void Meet(Animal other)
    {
        if (other == null)
        {
            Console.WriteLine("[Animal] Не с кем знакомиться.");
            return;
        }
        Console.WriteLine($"[Animal] {Name} знакомится с {other.Name}.");
        MakeSound();
        other.MakeSound();
    }

    // ==================== Взаимодействие: покормить другого Animal ====================
    public void Feed(Animal other, double amount)
    {
        if (other == null)
        {
            Console.WriteLine("[Animal] Некого кормить.");
            return;
        }
        Console.WriteLine($"[Animal] {Name} кормит {other.Name}.");
        other.Eat(amount);
    }
}

// ==================== Производный класс Dog ====================
public class Dog : Animal
{
    private string _breed;
    private int _energyLevel;

    public string Breed
    {
        get { return _breed; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Порода не может быть пустой!");
            _breed = value;
        }
    }

    public int EnergyLevel
    {
        get { return _energyLevel; }
        set
        {
            if (value < 0) _energyLevel = 0;
            else if (value > 100) _energyLevel = 100;
            else _energyLevel = value;
        }
    }

    public Dog() : base()
    {
        _breed = "Дворняга";
        _energyLevel = 100;
        Console.WriteLine("[Dog] Создана собака по умолчанию.");
    }

    public Dog(string name, int age, double weight, string breed)
        : base(name, age, weight)
    {
        Breed = breed;
        EnergyLevel = 100;
        Console.WriteLine($"[Dog] Создана собака: {Name}, порода: {Breed}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }

    // Взаимодействие: игра с другим Animal
    public void PlayWith(Animal other)
    {
        if (other == null)
        {
            Console.WriteLine("[Dog] Некого развлекать.");
            return;
        }
        EnergyLevel -= 15;
        Console.WriteLine($"[Dog] {Name} играет с {other.Name}. Энергия: {EnergyLevel}");
        other.MakeSound();
    }

    public void ShowDogInfo()
    {
        ShowInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}");
    }
}

// ==================== Производный класс Cat ====================
public class Cat : Animal
{
    private bool _isIndoor;
    private int _livesLeft;

    public bool IsIndoor
    {
        get { return _isIndoor; }
        set { _isIndoor = value; }
    }

    public int LivesLeft
    {
        get { return _livesLeft; }
        set
        {
            if (value < 0 || value > 9)
                throw new ArgumentOutOfRangeException("Жизней у кота от 0 до 9!");
            _livesLeft = value;
        }
    }

    public Cat() : base()
    {
        _isIndoor = true;
        _livesLeft = 9;
        Console.WriteLine("[Cat] Создан кот по умолчанию.");
    }

    public Cat(string name, int age, double weight, bool isIndoor)
        : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Console.WriteLine($"[Cat] Создан кот: {Name}, домашний: {IsIndoor}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }

    // Взаимодействие: кот «успокаивает» другое животное
    public void Calm(Animal other)
    {
        if (other == null)
        {
            Console.WriteLine("[Cat] Некого успокаивать.");
            return;
        }
        Console.WriteLine($"[Cat] {Name} успокаивает {other.Name}.");
        Purr();
    }

    public void ShowCatInfo()
    {
        ShowInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}");
    }
}

// ==================== Производный класс Parrot ====================
public class Parrot : Animal
{
    private string _phrase;
    private bool _canFly;

    public string Phrase
    {
        get { return _phrase; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Фраза не может быть пустой!");
            _phrase = value;
        }
    }

    public bool CanFly
    {
        get { return _canFly; }
        set { _canFly = value; }
    }

    public Parrot() : base()
    {
        _phrase = "Кеша хороший";
        _canFly = true;
        Console.WriteLine("[Parrot] Создан попугай по умолчанию.");
    }

    public Parrot(string name, int age, double weight, string phrase)
        : base(name, age, weight)
    {
        Phrase = phrase;
        CanFly = true;
        Console.WriteLine($"[Parrot] Создан попугай: {Name}, фраза: {Phrase}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Parrot] {Name} говорит: {Phrase}!");
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Parrot] {Name} летает!");
        else
            Console.WriteLine($"[Parrot] {Name} не умеет летать.");
    }

    // Взаимодействие: попугай «повторяет» за другим Animal
    public void RepeatAfter(Animal other)
    {
        if (other == null)
        {
            Console.WriteLine("[Parrot] Некого повторять.");
            return;
        }
        Console.WriteLine($"[Parrot] {Name} повторяет за {other.Name}:");
        other.MakeSound();
        MakeSound();
    }

    public void ShowParrotInfo()
    {
        ShowInfo();
        Console.WriteLine($"[Parrot] Фраза: {Phrase}, Умеет летать: {CanFly}");
    }
}

// ==================== Демонстрация (top-level код) ====================

Console.WriteLine("=== Создание экземпляров классов ===\n");

Animal genericAnimal = new Animal("Зверь", 3, 10.5);
Dog dog = new Dog("Рекс", 5, 25.0, "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, true);
Parrot parrot = new Parrot("Кеша", 1, 0.4, "Кеша хороший");

Console.WriteLine("\n=== Полиморфизм: MakeSound ===\n");
genericAnimal.MakeSound();
dog.MakeSound();
cat.MakeSound();
parrot.MakeSound();

Console.WriteLine("\n=== Свойства и состояние ===\n");
dog.ShowDogInfo();
cat.ShowCatInfo();
parrot.ShowParrotInfo();

Console.WriteLine("\n=== Изменение свойств через сеттеры ===\n");
dog.EnergyLevel = 50;
cat.LivesLeft = 7;
parrot.Phrase = "Попка дурак";

Console.WriteLine($"dog.EnergyLevel = {dog.EnergyLevel}");
Console.WriteLine($"cat.LivesLeft = {cat.LivesLeft}");
Console.WriteLine($"parrot.Phrase = {parrot.Phrase}");

Console.WriteLine("\n=== Взаимодействие объектов между собой ===\n");

// 1. Знакомство: dog знакомится с cat
dog.Meet(cat);

Console.WriteLine();

// 2. Кормление: dog кормит parrot
dog.Feed(parrot, 5.0);

Console.WriteLine();

// 3. Игра: dog играет с cat
dog.PlayWith(cat);

Console.WriteLine();

// 4. Кот успокаивает собаку
cat.Calm(dog);

Console.WriteLine();

// 5. Попугай повторяет за котом
parrot.RepeatAfter(cat);

Console.WriteLine();

// 6. Собака приносит тапочки (собственный метод Dog)
dog.BringSlippers();

Console.WriteLine();

// 7. Попугай летает (собственный метод Parrot)
parrot.Fly();

Console.WriteLine("\n=== Проверка валидации сеттеров ===\n");
try { cat.LivesLeft = 15; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { dog.Name = ""; }
catch (ArgumentException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

try { genericAnimal.Age = -5; }
catch (ArgumentOutOfRangeException ex) { Console.WriteLine($"Ошибка: {ex.Message}"); }

Console.WriteLine("\n=== Статические члены ===\n");
Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals (через свойство): {Animal.TotalAnimals}");
````
