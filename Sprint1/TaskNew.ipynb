```
using System;
using System.Collections.Generic;

//1. Товар (Product)
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value;}
    }

    public int Quantity
    {
        get { return quantity; }
        set{quantity = value;}
    }

    public Product(string name, decimal price, int quantity)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        category = "Без категории";
        description = "Без описания";
    }

    public decimal GetTotalPrice() //чтобы вернуть общую стоимость всех единиц товара
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount) //обновляем кол-во тов на указанное значение
    {
        Quantity = amount;
        Console.WriteLine($"[Product] {Name}: количество обновлено = {Quantity}");
    }

    public string GetProductInfo()
    {
        return $"Товар: {Name}, Цена: {Price:C}, Количество: {Quantity}, Категория: {category}, Описание: {description}";
    }
}

//2.Продавец (Seller)
public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set
        { salary = value; }
    }

    public Seller(string name, string employeeId, decimal salary)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        contactInfo = "Не указан";
    }

    public void AddProduct(Product product) //в список товаров продавца
    {
        if (product == null)
        {
            Console.WriteLine("[Seller] Нечего добавлять.");
            return;
        }
        products.Add(product);
        Console.WriteLine($"[Seller] {Name} добавил товар: {product.Name}");
    }

    public void SellProduct(Product product, int quantity) // -товар на складе
    {
        if (product == null)
        {
            Console.WriteLine("[Seller] Товар не найден.");
            return;
        }
        if (quantity <= 0 || quantity > product.Quantity)
        {
            Console.WriteLine($"[Seller] Недостаточно товара «{product.Name}» на складе ({product.Quantity}).");
            return;
        }
        product.UpdateQuantity(product.Quantity - quantity);
        Console.WriteLine($"[Seller] {Name} продал {quantity} шт. товара «{product.Name}»");
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {Name}, ID: {EmployeeId}, Зарплата: {Salary:C}, Контакт: {contactInfo}, Товаров: {products.Count}";
    }
}

//3Магазин(Store)
public class Store
{
    private string storeName;
    private string location;
    public string storeHours;

    private List<Seller> sellers = new List<Seller>();
    private List<Product> products = new List<Product>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller) // проавца в магаз
    {
        sellers.Add(seller);
        Console.WriteLine($"[Store] {seller.Name} принят на работу в «{StoreName}»");
    }

    public void AddProduct(Product product)
    {
        if (product == null) return;
        products.Add(product);
    }

    public void ListProducts() //перечислениее тов в маг
    {
        Console.WriteLine($"\n[Store] Товары в магазине «{StoreName}»:");
        if (products.Count == 0)
        {
            Console.WriteLine("  (пусто)");
            return;
        }
        foreach (var p in products)
        {
            Console.WriteLine($"  {p.GetProductInfo()}");
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {StoreName}, Адрес: {Location}, Часы работы: {storeHours}, Продавцов: {sellers.Count}, Товаров: {products.Count}";
    }
}

Product p1 = new Product("Ноутбук", 50000m, 10);
Product p2 = new Product("Телефон", 30000m, 5);
Product p3 = new Product("Планшет", 25000m, 8);

Seller s1 = new Seller("Игорь", "EMP001", 60000m);
Seller s2 = new Seller("Ира", "EMP002", 55000m);

Store store = new Store("Мвидео", "ул. Ленина, 10", "09:00 - 21:00");


s1.AddProduct(p1);
s1.AddProduct(p2); //тов к прод
s2.AddProduct(p3);

store.AddSeller(s1); //прод к маг
store.AddSeller(s2);

store.AddProduct(p1);
store.AddProduct(p2); //тов к маг
store.AddProduct(p3);

Console.WriteLine(s1.GetSellerInfo());
Console.WriteLine(s2.GetSellerInfo());

s1.SellProduct(p1, 2);   //продажа
s2.SellProduct(p3, 3);  

Console.WriteLine($"{p1.Name}: {p1.GetTotalPrice():C}");
Console.WriteLine($"{p2.Name}: {p2.GetTotalPrice():C}"); //общ стоим тов
Console.WriteLine($"{p3.Name}: {p3.GetTotalPrice():C}");

Console.WriteLine(store.GetStoreInfo());

store.ListProducts();
```
