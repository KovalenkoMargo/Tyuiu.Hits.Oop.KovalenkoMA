```
using System;
using System.Collections.Generic;

public abstract class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }
    public string Color { get; set; }
    public bool IsWild { get; set; }
    public int Id { get; set; }

    public static int TotalAnimals { get; private set; } = 0;

    protected Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
        Habitat = "Неизвестно";
        Color = "Неизвестный";
        IsWild = true;
        Id = TotalAnimals;
    }

    protected Animal(string name, int age, double weight, string habitat) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    // перекрытие
    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[Animal] #{Id} {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Среда: {Habitat}, Окрас: {Color}, Дикое: {IsWild}");
    }

    // перегрузка
    public void Eat(double amount)
    {
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел {amount} кг. Вес: {Weight:F2} кг");
    }

    public void Eat(string food)
    {
        Weight += 0.2;
        Console.WriteLine($"[Animal] {Name} поел «{food}». Вес: {Weight:F2} кг");
    }

    public void Eat(double amount, string food)
    {
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел {amount} кг «{food}». Вес: {Weight:F2} кг");
    }

    public void Move(int distance)
    {
        Console.WriteLine($"[Animal] {Name} перемещается на {distance} м.");
    }

    public void Move(int distance, string mode)
    {
        Console.WriteLine($"[Animal] {Name} перемещается на {distance} м способом «{mode}».");
    }

    public void Move(string direction)
    {
        Console.WriteLine($"[Animal] {Name} перемещается в направлении «{direction}».");
    }

    public void Migrate(string newHabitat)
    {
        Console.WriteLine($"[Animal] {Name} мигрирует: {Habitat} → {newHabitat}");
        Habitat = newHabitat;
    }

    public void Sleep(int hours)
    {
        Weight += hours * 0.01;
        Console.WriteLine($"[Animal] {Name} спал {hours} ч. Вес: {Weight:F2} кг");
    }

    public void Play()
    {
        Console.WriteLine($"[Animal] {Name} играет.");
    }

    public void Rest()
    {
        Console.WriteLine($"[Animal] {Name} отдыхает.");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }

    public void Meet(Animal other)
    {
        Console.WriteLine($"[Animal] {Name} знакомится с {other.Name}.");
        MakeSound();
        other.MakeSound();
    }
}


public interface IPet
{
    string Owner { get; set; }
    void ShowAffection();
}

public interface IFlyable
{
    bool CanFly { get; set; }
    void Fly();
}

public interface IHunter
{
    void Hunt();
}

public class Mammal : Animal
{
    public string FurColor { get; set; }
    public int LegsCount { get; set; }

    public bool HasTail { get; set; }
    public string DietType { get; set; }
    public double BodyTemp { get; set; }

    public Mammal(string name, int age, double weight, string habitat,
                  string furColor, int legsCount)
        : base(name, age, weight, habitat)
    {
        FurColor = furColor;
        LegsCount = legsCount;
        HasTail = true;
        DietType = "Всеядный";
        BodyTemp = 37.0;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Mammal] Шерсть: {FurColor}, Ног: {LegsCount}, Хвост: {HasTail}, Питание: {DietType}");
    }

    public void Groom()
    {
        Console.WriteLine($"[Mammal] {Name} вычёсывает шерсть ({FurColor}).");
    }

    public void Groom(string brush)
    {
        Console.WriteLine($"[Mammal] {Name} вычёсывает шерсть щёткой «{brush}».");
    }

    public void Walk()
    {
        Console.WriteLine($"[Mammal] {Name} ходит на {LegsCount} ногах.");
    }

    public void Walk(int steps)
    {
        Console.WriteLine($"[Mammal] {Name} прошёл {steps} шагов.");
    }

    public void FeedMilk()
    {
        Console.WriteLine($"[Mammal] {Name} кормит детёныша молоком.");
    }

    public void Hibernate()
    {
        Console.WriteLine($"[Mammal] {Name} впадает в спячку.");
    }

    public void ShedFur()
    {
        Console.WriteLine($"[Mammal] {Name} линяет ({FurColor}).");
    }

    public void RegulateTemp()
    {
        Console.WriteLine($"[Mammal] {Name} поддерживает t° = {BodyTemp}°C.");
    }
}


public class Bird : Animal, IFlyable, IPet
{
    public string Species { get; set; }
    public double Wingspan { get; set; }
    public string FeatherColor { get; set; }
    public bool CanFly { get; set; }
    public string Owner { get; set; }

    public int EggsPerYear { get; set; }
    public bool IsMigratory { get; set; }
    public string BeakType { get; set; }

    public Bird(string name, int age, double weight, string habitat,
                string species, double wingspan, string featherColor, bool canFly)
        : base(name, age, weight, habitat)
    {
        Species = species;
        Wingspan = wingspan;
        FeatherColor = featherColor;
        CanFly = canFly;
        Owner = "Без хозяина";
        EggsPerYear = 5;
        IsMigratory = false;
        BeakType = "Короткий";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Bird] Вид: {Species}, Размах: {Wingspan} м, Перья: {FeatherColor}, Летает: {CanFly}, Яиц/год: {EggsPerYear}");
    }

    //  ЯВНАЯ РЕАЛИЗАЦИЯ IFlyable =
    // Доступно только через IFlyable
    void IFlyable.Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает с размахом {Wingspan} м!");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }

    public void FlyPublic()
    {
        ((IFlyable)this).Fly();
    }

    //  ЯВНАЯ РЕАЛИЗАЦИЯ IPet 
    void IPet.ShowAffection()
    {
        Console.WriteLine($"[Bird] {Name} сидит на плече у {Owner}.");
    }

    public void ShowAffectionPublic()
    {
        ((IPet)this).ShowAffection();
    }

    // перегрузка
    public void Fly(int height)
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает на высоте {height} м.");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }

    public void Sing()
    {
        Console.WriteLine($"[Bird] {Name} поёт.");
    }

    public void Sing(string song)
    {
        Console.WriteLine($"[Bird] {Name} поёт «{song}».");
    }

    public void LayEggs()
    {
        Console.WriteLine($"[Bird] {Name} откладывает яйца ({EggsPerYear} шт/год).");
    }

    public void MigrateSeason()
    {
        if (IsMigratory)
            Console.WriteLine($"[Bird] {Name} мигрирует на юг.");
        else
            Console.WriteLine($"[Bird] {Name} остаётся зимовать.");
    }

    public void Peck()
    {
        Console.WriteLine($"[Bird] {Name} клюёт корм.");
    }

    public void BuildNest()
    {
        Console.WriteLine($"[Bird] {Name} вьёт гнездо.");
    }

    public void Preen()
    {
        Console.WriteLine($"[Bird] {Name} чистит перья ({FeatherColor}).");
    }
}

public class Dog : Mammal, IPet, IHunter
{
    public string Breed { get; set; }
    public int EnergyLevel { get; set; }
    public string Owner { get; set; }

    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public int BarkVolume { get; set; }

    public Dog(string name, int age, double weight, string habitat,
               string furColor, int legsCount, string breed, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        Breed = breed;
        EnergyLevel = 100;
        Owner = owner;
        IsTrained = false;
        FavoriteToy = "Мяч";
        BarkVolume = 70;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Dog] Порода: {Breed}, Энергия: {EnergyLevel}, Хозяин: {Owner}, Дрессирован: {IsTrained}, Игрушка: {FavoriteToy}");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IHunter 
    void IHunter.Hunt()
    {
        EnergyLevel -= 20;
        Console.WriteLine($"[Dog] {Name} охотится! Энергия: {EnergyLevel}");
    }

    // НЕЯВНАЯ РЕАЛИЗАЦИЯ IPet 
    public void ShowAffection()
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} виляет хвостом перед {Owner}. Энергия: {EnergyLevel}");
    }

    // перегрузка
    public void Bark()
    {
        Console.WriteLine($"[Dog] {Name} лает: ГАВ!");
    }

    public void Bark(int times)
    {
        Console.Write($"[Dog] {Name} лает {times} раз: ");
        for (int i = 0; i < times; i++) Console.Write("ГАВ! ");
        Console.WriteLine();
    }

    public void Bark(int times, int volume)
    {
        BarkVolume = volume;
        Console.WriteLine($"[Dog] {Name} лает {times} раз с громкостью {volume}.");
    }

    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }

    public void BringSlippers(string item)
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} принёс «{item}»! Энергия: {EnergyLevel}");
    }

    public void Train()
    {
        IsTrained = true;
        Console.WriteLine($"[Dog] {Name} теперь дрессирован!");
    }

    public void PlayWithToy()
    {
        EnergyLevel -= 8;
        Console.WriteLine($"[Dog] {Name} играет с «{FavoriteToy}». Энергия: {EnergyLevel}");
    }

    public void GuardHouse()
    {
        EnergyLevel -= 15;
        Console.WriteLine($"[Dog] {Name} охраняет дом {Owner}. Энергия: {EnergyLevel}");
    }
}

public class GuideDog : Dog
{
    public string OwnerName { get; set; }
    public bool IsWorking { get; set; }
    public string HarnessColor { get; set; }

    public int YearsOfService { get; set; }
    public string Certification { get; set; }
    public bool KnowsCommands { get; set; }

    public GuideDog(string name, int age, double weight, string habitat,
                    string furColor, int legsCount, string breed, string owner,
                    string ownerName, bool isWorking, string harnessColor)
        : base(name, age, weight, habitat, furColor, legsCount, breed, owner)
    {
        OwnerName = ownerName;
        IsWorking = isWorking;
        HarnessColor = harnessColor;
        YearsOfService = 1;
        Certification = "Базовая";
        KnowsCommands = true;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[GuideDog] Владелец: {OwnerName}, На работе: {IsWorking}, Упряжь: {HarnessColor}, Лет службы: {YearsOfService}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[GuideDog] {Name} (поводырь) тихо гавкает: Гав!");
    }

    public void Guide()
    {
        if (IsWorking)
            Console.WriteLine($"[GuideDog] {Name} ведёт {OwnerName}.");
        else
            Console.WriteLine($"[GuideDog] {Name} отдыхает.");
    }

    public void WearHarness()
    {
        Console.WriteLine($"[GuideDog] {Name} надевает {HarnessColor} упряжь.");
    }

    public void WearHarness(string color)
    {
        HarnessColor = color;
        Console.WriteLine($"[GuideDog] {Name} надевает упряжь цвета «{color}».");
    }

    public void Retire()
    {
        IsWorking = false;
        Console.WriteLine($"[GuideDog] {Name} вышел на пенсию после {YearsOfService} лет службы.");
    }

    public void GetCertified(string level)
    {
        Certification = level;
        Console.WriteLine($"[GuideDog] {Name} получил сертификат «{level}».");
    }

    public void CrossRoad()
    {
        Console.WriteLine($"[GuideDog] {Name} переводит {OwnerName} через дорогу.");
    }

    public void StopWorking()
    {
        IsWorking = false;
        Console.WriteLine($"[GuideDog] {Name} закончил работу.");
    }

    public void CheckHarness()
    {
        Console.WriteLine($"[GuideDog] {Name} проверяет упряжь «{HarnessColor}».");
    }
}

public class Cat : Mammal, IPet, IHunter
{
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }
    public string Owner { get; set; }

    public bool IsNeutered { get; set; }
    public string FavoriteFood { get; set; }
    public int PurrVolume { get; set; }

    public Cat(string name, int age, double weight, string habitat,
               string furColor, int legsCount, bool isIndoor, string owner)
        : base(name, age, weight, habitat, furColor, legsCount)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Owner = owner;
        IsNeutered = false;
        FavoriteFood = "Рыба";
        PurrVolume = 50;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Cat] Домашний: {IsIndoor}, Жизней: {LivesLeft}, Хозяин: {Owner}, Стерилизован: {IsNeutered}, Еда: {FavoriteFood}");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IHunter 
    void IHunter.Hunt()
    {
        Console.WriteLine($"[Cat] {Name} охотится на мышей!");
    }

    //  ЯВНАЯ РЕАЛИЗАЦИЯ IPet 
    void IPet.ShowAffection()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет рядом с {Owner}.");
    }

    public void ShowAffectionPublic()
    {
        ((IPet)this).ShowAffection();
    }

    // перегрузка
    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }

    public void Purr(int minutes)
    {
        Console.WriteLine($"[Cat] {Name} мурлычет {minutes} минут...");
    }

    public void Purr(int minutes, int volume)
    {
        PurrVolume = volume;
        Console.WriteLine($"[Cat] {Name} мурлычет {minutes} мин с громкостью {volume}.");
    }

    public void EatFavorite()
    {
        Console.WriteLine($"[Cat] {Name} ест любимую еду «{FavoriteFood}».");
    }

    public void Scratch()
    {
        Console.WriteLine($"[Cat] {Name} точит когти.");
    }

    public void ClimbTree()
    {
        Console.WriteLine($"[Cat] {Name} залезла на дерево.");
    }
}

public class Parrot : Bird
{
    public string Phrase { get; set; }
    public int VocabularySize { get; set; }

    public string BeakColor { get; set; }
    public bool CanTalk { get; set; }
    public int LifespanYears { get; set; }

    public Parrot(string name, int age, double weight, string habitat,
                  string species, double wingspan, string featherColor, bool canFly,
                  string phrase, int vocabularySize)
        : base(name, age, weight, habitat, species, wingspan, featherColor, canFly)
    {
        Phrase = phrase;
        VocabularySize = vocabularySize;
        BeakColor = "Жёлтый";
        CanTalk = true;
        LifespanYears = 50;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"[Parrot] Фраза: «{Phrase}», Словарь: {VocabularySize}, Клюв: {BeakColor}, Говорит: {CanTalk}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Parrot] {Name} говорит: {Phrase}!");
    }

    // перегрузка
    public void Repeat(string words)
    {
        Console.WriteLine($"[Parrot] {Name} повторяет: {words}");
    }

    public void Repeat(string words, int times)
    {
        for (int i = 0; i < times; i++)
            Console.WriteLine($"[Parrot] {Name} повторяет: {words}");
    }

    public void LearnWord(string word)
    {
        VocabularySize++;
        Console.WriteLine($"[Parrot] {Name} выучил слово «{word}». Словарь: {VocabularySize}");
    }

    public void LearnWord(string word, int repeat)
    {
        VocabularySize += repeat;
        Console.WriteLine($"[Parrot] {Name} выучил слово «{word}» ({repeat} раз). Словарь: {VocabularySize}");
    }

    public void ChangeBeakColor(string color)
    {
        BeakColor = color;
        Console.WriteLine($"[Parrot] {Name} сменил цвет клюва на «{color}».");
    }

    public void ShowLifespan()
    {
        Console.WriteLine($"[Parrot] {Name} может прожить до {LifespanYears} лет.");
    }

    public void Dance()
    {
        Console.WriteLine($"[Parrot] {Name} танцует.");
    }
}

public class AnimalCollection<T> where T : Animal
{
    private List<T> _animals = new List<T>();

    public void Add(T animal) => _animals.Add(animal);
    public void Remove(T animal) => _animals.Remove(animal);
    public int Count => _animals.Count;

    public void DisplayAll()
    {
        foreach (var a in _animals)
        {
            a.DisplayInfo();
            a.MakeSound();
            Console.WriteLine();
        }
    }
}


// Сервис для любых животных — зависит от Animal (абстрактный класс)
public class AnimalService
{
    private readonly Animal _animal;

    public AnimalService(Animal animal)
    {
        _animal = animal;
    }

    public void Interact()
    {
        Console.WriteLine("/AnimalService/");
        _animal.MakeSound();
        _animal.DisplayInfo();
        Console.WriteLine();
    }
}

// Сервис для домашних — зависит от IPet
public class PetService
{
    private readonly IPet _pet;

    public PetService(IPet pet)
    {
        _pet = pet;
    }

    public void Care()
    {
        Console.WriteLine("/PetService/");
        _pet.ShowAffection();
        Console.WriteLine();
    }
}

// Сервис для охотников — зависит от IHunter
public class HunterService
{
    private readonly IHunter _hunter;

    public HunterService(IHunter hunter)
    {
        _hunter = hunter;
    }

    public void SendToHunt()
    {
        Console.WriteLine("/HunterService/");
        _hunter.Hunt();
        Console.WriteLine();
    }
}

// Сервис для летающих — зависит от IFlyable
public class FlightService
{
    private readonly IFlyable _flyable;

    public FlightService(IFlyable flyable)
    {
        _flyable = flyable;
    }

    public void TestFlight()
    {
        Console.WriteLine("/FlightService/");
        _flyable.Fly();
        Console.WriteLine();
    }
}


Dog dog = new Dog("Рекс", 5, 25.0, "Дом", "Рыжий", 4, "Лабрадор", "Иван");
GuideDog guideDog = new GuideDog("Бадди", 4, 30.0, "Дом", "Золотистый", 4, "Лабрадор", "Иван", "Иван", true, "Красная");
Cat cat = new Cat("Мурка", 2, 4.2, "Дом", "Серый", 4, true, "Мария");
Bird bird = new Bird("Кеша", 1, 0.4, "Лес", "Попугай", 0.3, "Зелёный", true);
Parrot parrot = new Parrot("Гоша", 2, 0.5, "Дом", "Попугай", 0.35, "Красный", true, "Гоша хороший", 15);
```
