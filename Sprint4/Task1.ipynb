<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

````
using System;

public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса — доступна только через IVehicle
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }

    public void Accelerate(int amount)
    {
        currentSpeed += amount;
        Console.WriteLine($"Автомобиль разгоняется до {currentSpeed} км/ч.");
    }
}

// Абстрактный класс + производные 
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();
    public abstract void Stop();

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}

public class SportsCar : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} (спорткар) запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} (спорткар) остановлен.");
    }
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}


Console.WriteLine("ЯВНАЯ реализация интерфейса в Car \n");

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start();
myCar.Stop();
Console.WriteLine($"Скорость через интерфейс: {myCar.GetCurrentSpeed()}");

// Через экземпляр Car методы Start/Stop НЕ доступны:
// Car carDirect = new Car();
// carDirect.Start();   // ← ОШИБКА компиляции CS1061

Console.WriteLine("\n Абстрактный класс + производные\n");

Vehicle sportsCar = new SportsCar { Make = "Ferrari", Model = "488" };
sportsCar.Start();
sportsCar.Stop();
Console.WriteLine($"Скорость: {sportsCar.GetCurrentSpeed()}");

Console.WriteLine();

Motorcycle moto = new Motorcycle { Make = "Harley-Davidson", Model = "Street" };
moto.Start();
moto.RevEngine();
moto.Stop();

Console.WriteLine("\nНаследование интерфейсов\n");

IMotorizedVehicle motorized = moto;
motorized.Start();
motorized.RevEngine();
motorized.Stop();

Console.WriteLine("\nПолиморфизм через IVehicle[]\n");

IVehicle[] vehicles =
{
    new Car { Make = "BMW", Model = "X5" },
    new SportsCar { Make = "Porsche", Model = "911" },
    new Motorcycle { Make = "Yamaha", Model = "R1" }
};

foreach (IVehicle v in vehicles)
{
    v.Start();
    Console.WriteLine($"Скорость: {v.GetCurrentSpeed()}");
    v.Stop();
    Console.WriteLine();
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также применить явнаю реализацию интерфейса.

````
using System;

public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
}

public interface IPet
{
    string Owner { get; set; }
    void ShowAffection();
}

public interface IFlyable
{
    bool CanFly { get; set; }
    void Fly();
}

public interface IHunter
{
    void Hunt();
}

public interface IPetHunter : IPet, IHunter
{
    void TrackPrey();
}

//Абстрактный класс Animal 
public abstract class Animal : IAnimal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public static int TotalAnimals { get; private set; } = 0;

    protected Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
    }

    protected Animal(string name, int age, double weight) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void Move()
    {
        Console.WriteLine($"[Animal] {Name} перемещается.");
    }

    public virtual string GetInfo()
    {
        return $"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг";
    }

    public void Eat(double amount)
    {
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal, IPetHunter
{
    public string Breed { get; set; }
    public int EnergyLevel { get; set; }

    // IPet — неявная реализация
    public string Owner { get; set; }

    public Dog(string name, int age, double weight, string breed, string owner)
        : base(name, age, weight)
    {
        Breed = breed;
        EnergyLevel = 100;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void Move()
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} бежит на четырёх лапах. Энергия: {EnergyLevel}");
    }

    public override string GetInfo()
    {
        return $"[Dog] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Порода: {Breed}, Хозяин: {Owner}, Энергия: {EnergyLevel}";
    }

    // IPet — неявная
    public void ShowAffection()
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} виляет хвостом перед {Owner}. Энергия: {EnergyLevel}");
    }

    // IHunter — ЯВНАЯ реализация (доступна только через IHunter)
    void IHunter.Hunt()
    {
        EnergyLevel -= 20;
        Console.WriteLine($"[Dog] {Name} охотится! Энергия: {EnergyLevel}");
    }

    // IPetHunter — ЯВНАЯ реализация
    void IPetHunter.TrackPrey()
    {
        EnergyLevel -= 15;
        Console.WriteLine($"[Dog] {Name} выслеживает добычу! Энергия: {EnergyLevel}");
    }

    // Собственный метод
    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }
}

public class Cat : Animal, IPetHunter
{
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }

    // IPet — неявная
    public string Owner { get; set; }

    public Cat(string name, int age, double weight, bool isIndoor, string owner)
        : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void Move()
    {
        Console.WriteLine($"[Cat] {Name} грациозно идёт. Жизней: {LivesLeft}");
    }

    public override string GetInfo()
    {
        return $"[Cat] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Домашний: {IsIndoor}, Жизней: {LivesLeft}, Хозяин: {Owner}";
    }

    // IPet — неявная
    public void ShowAffection()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет рядом с {Owner}.");
    }

    // IHunter — неявная
    public void Hunt()
    {
        Console.WriteLine($"[Cat] {Name} охотится на мышей!");
    }

    // IPetHunter — ЯВНАЯ реализация
    void IPetHunter.TrackPrey()
    {
        Console.WriteLine($"[Cat] {Name} крадётся за добычей...");
    }

    // Собственный метод
    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }
}

// ==================== Bird — явная реализация IFlyable ====================
public class Bird : Animal, IFlyable, IPet
{
    public double Wingspan { get; set; }

    // IFlyable — неявная
    public bool CanFly { get; set; }

    // IPet — неявная
    public string Owner { get; set; }

    public Bird(string name, int age, double weight, double wingspan, bool canFly)
        : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        Owner = "Без хозяина";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void Move()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летит с размахом {Wingspan} м.");
        else
            Console.WriteLine($"[Bird] {Name} идёт пешком.");
    }

    public override string GetInfo()
    {
        return $"[Bird] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг, Размах: {Wingspan} м, Летает: {CanFly}, Хозяин: {Owner}";
    }

    // IFlyable — ЯВНАЯ реализация
    void IFlyable.Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает!");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }

    // IPet — неявная
    public void ShowAffection()
    {
        Console.WriteLine($"[Bird] {Name} сидит на плече у {Owner}.");
    }

    public void Sing()
    {
        Console.WriteLine($"[Bird] {Name} поёт!");
    }

    public void FlyPublic()
    {
        ((IFlyable)this).Fly();
    }
}

Dog dog = new Dog("Рекс", 5, 25.0, "Лабрадор", "Иван");
Cat cat = new Cat("Мурка", 2, 4.2, true, "Мария");
Bird bird = new Bird("Кеша", 1, 0.4, 0.3, true);

Console.WriteLine(" через IAnimal \n");

IAnimal[] animals = { dog, cat, bird };
foreach (IAnimal a in animals)
{
    a.MakeSound();
    a.Move();
    Console.WriteLine(a.GetInfo());
    Console.WriteLine();
}
````

````
Console.WriteLine("через IPet (неявная реализация)\n");

IPet[] pets = { dog, cat, bird };
foreach (IPet p in pets)
    p.ShowAffection();

Console.WriteLine("\nчерез IHunter\n");

// У Dog метод Hunt реализован ЯВНО → доступен только через IHunter
IHunter hunterDog = dog;
hunterDog.Hunt();

// У Cat метод Hunt реализован НЕЯВНО → доступен и так, и через интерфейс
cat.Hunt();

Console.WriteLine("\nчерез IPetHunter\n");

IPetHunter petHunterDog = dog;
petHunterDog.TrackPrey();
petHunterDog.ShowAffection();
petHunterDog.Hunt();           // через IPetHunter (унаследован от IHunter)

Console.WriteLine();

IPetHunter petHunterCat = cat;
petHunterCat.TrackPrey();      // явная реализация
petHunterCat.Hunt();           // неявная реализация
````

````
Console.WriteLine("\nчерез IFlyable\n");

// У Bird метод Fly реализован ЯВНО → доступен только через IFlyable
IFlyable flyableBird = bird;
flyableBird.Fly();

bird.FlyPublic();

Console.WriteLine("\nЯВНАЯ реализация: попытка вызвать напрямую\n");

Console.WriteLine("Методы Hunt (Dog), Fly (Bird), TrackPrey — доступны только через интерфейсы.");

Console.WriteLine("\nСобственные методы классов\n");

dog.BringSlippers();
cat.Purr();
bird.Sing();

Console.WriteLine("\n Изменение свойств через сеттеры\n");

dog.EnergyLevel = 50;
cat.LivesLeft = 7;
bird.CanFly = false;

Console.WriteLine($"dog.EnergyLevel = {dog.EnergyLevel}");
Console.WriteLine($"cat.LivesLeft = {cat.LivesLeft}");
Console.WriteLine($"bird.CanFly = {bird.CanFly}");

Animal.DisplayTotalAnimals();
Console.WriteLine($"TotalAnimals: {Animal.TotalAnimals}");
````
