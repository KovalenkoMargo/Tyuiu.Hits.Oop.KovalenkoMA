### Интерфейсы в C#

**Интерфейс** — это контракт, который определяет набор методов и свойств, которые класс должен реализовать. Интерфейсы позволяют создать гибкую и расширяемую архитектуру, поскольку они отделяют определение поведения от его реализации.

Давайте подробно разберем понятия интерфейсов и управления зависимостями в C# на примере классов `Vehicle`, `Car` и `Motorcycle`.

#### Пример интерфейса

Предположим, мы имеем интерфейс `IVehicle`, который определяет поведение транспортных средств. Ниже приведен пример его определения:

```csharp
public interface IVehicle
{
    void Start();            // Метод для запуска транспортного средства
    void Stop();             // Метод для остановки транспортного средства
    int GetNumberOfWheels(); // Метод для получения количества колес
}
```

Интерфейс `IVehicle` требует, чтобы любой класс, реализующий его, обеспечил реализацию этих методов.

### Реализация интерфейса

Теперь создадим классы `Car` и `Motorcycle`, которые будут реализовывать этот интерфейс:

```csharp
public class Car : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Машина запущена.");
    }

    public void Stop()
    {
        Console.WriteLine("Машина остановлена.");
    }

    public int GetNumberOfWheels()
    {
        return 4; // У машины 4 колеса
    }
}

public class Motorcycle : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Мотоцикл запущен.");
    }

    public void Stop()
    {
        Console.WriteLine("Мотоцикл остановлен.");
    }

    public int GetNumberOfWheels()
    {
        return 2; // У мотоцикла 2 колеса
    }
}
```

Каждый класс (`Car` и `Motorcycle`) реализует все методы, определенные в интерфейсе `IVehicle`, что позволяет создать единый интерфейс для работы с разными типами транспортных средств.

### Управление зависимостями (Dependency Injection)

**Управление зависимостями** — это паттерн проектирования, который помогает управлять созданием зависимостей в классе. Этот подход позволяет внедрять (или "вводить") зависимости через конструктор, методы или свойства, вместо того чтобы создавать их внутри класса. Это делает код более чистым и облегчает его тестирование.

#### Пример управления зависимостями

Создадим класс `VehicleService`, который будет зависеть от интерфейса `IVehicle`:

```csharp
public class VehicleService
{
    private readonly IVehicle _vehicle;

    // Внедрение зависимости через конструктор
    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle; 
    }

    public void UseVehicle()
    {
        _vehicle.Start(); // Запускаем транспортное средство
        Console.WriteLine($"У этого транспортного средства {_vehicle.GetNumberOfWheels()} колеса.");
        _vehicle.Stop(); // Останавливаем транспортное средство
    }
}
```

В этом примере класс `VehicleService` принимает в своем конструкторе объект, реализующий интерфейс `IVehicle`. Это позволяет `VehicleService` работать с любым объектом, который реализует `IVehicle`, обеспечивая гибкость и уменьшая связывание между классами.

### Использование классов

Теперь давайте создадим объекты `Car` и `Motorcycle` и используем наш сервис:

```csharp
IVehicle myCar = new Car(); // Создаем объект машины
VehicleService carService = new VehicleService(myCar); // Передаем машину в сервис
carService.UseVehicle(); // Используем сервис с машиной

IVehicle myMotorcycle = new Motorcycle(); // Создаем объект мотоцикла
VehicleService motorcycleService = new VehicleService(myMotorcycle); // Передаем мотоцикл в сервис
motorcycleService.UseVehicle(); // Используем сервис с мотоциклом

```

### Результат выполнения

При запуске данной программы будет следующий вывод:

```
Машина запущена.
У этого транспортного средства 4 колеса.
Машина остановлена.
Мотоцикл запущен.
У этого транспортного средства 2 колеса.
Мотоцикл остановлен.
```

### Заключение

Использование интерфейсов и управления зависимостями позволяет создать более организованный и тестируемый код. Интерфейсы определяют стандартные поведения, а управление зависимостями снижает связность между компонентами. Это делает систему более гибкой и упрощает процесс её поддержки и расширения в будущем.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке с применением интерфейса и абстактного класса:</h4>

````
using System;

public interface IVehicle
{
    void Start();
    void Stop();
    int GetNumberOfWheels();
}

public class Car : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Машина запущена.");
    }

    public void Stop()
    {
        Console.WriteLine("Машина остановлена.");
    }

    public int GetNumberOfWheels()
    {
        return 4;
    }
}

public class Motorcycle : IVehicle
{
    public void Start()
    {
        Console.WriteLine("Мотоцикл запущен.");
    }

    public void Stop()
    {
        Console.WriteLine("Мотоцикл остановлен.");
    }

    public int GetNumberOfWheels()
    {
        return 2;
    }
}

public class VehicleService
{
    private readonly IVehicle _vehicle;

    // Внедрение зависимости через конструктор
    public VehicleService(IVehicle vehicle)
    {
        _vehicle = vehicle;
    }

    public void UseVehicle()
    {
        _vehicle.Start();
        Console.WriteLine($"У этого транспортного средства {_vehicle.GetNumberOfWheels()} колеса.");
        _vehicle.Stop();
    }
}


Console.WriteLine(" Прямое использование через интерфейс\n");

IVehicle myCar = new Car();
myCar.Start();
Console.WriteLine($"Колёс: {myCar.GetNumberOfWheels()}");
myCar.Stop();

Console.WriteLine();

IVehicle myMotorcycle = new Motorcycle();
myMotorcycle.Start();
Console.WriteLine($"Колёс: {myMotorcycle.GetNumberOfWheels()}");
myMotorcycle.Stop();

Console.WriteLine("\n Управление зависимостями через VehicleService\n");

VehicleService carService = new VehicleService(myCar);
carService.UseVehicle();

Console.WriteLine();

VehicleService motorcycleService = new VehicleService(myMotorcycle);
motorcycleService.UseVehicle();

Console.WriteLine("\n Один сервис — разные зависимости\n");

IVehicle[] vehicles = { new Car(), new Motorcycle(), new Car() };

foreach (IVehicle v in vehicles)
{
    VehicleService service = new VehicleService(v);
    service.UseVehicle();
    Console.WriteLine();
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также реализовать управление зависимостями.

````
using System;

public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
}

public interface IPet
{
    string Owner { get; set; }
    void ShowAffection();
}

public interface IFlyable
{
    bool CanFly { get; set; }
    void Fly();
}

public interface IHunter
{
    void Hunt();
}

public abstract class Animal : IAnimal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public static int TotalAnimals { get; private set; } = 0;

    protected Animal()
    {
        TotalAnimals++;
        Name = "Безымянное";
        Age = 0;
        Weight = 1.0;
    }

    protected Animal(string name, int age, double weight) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"[Animal] {Name} издаёт неопределённый звук.");
    }

    public virtual void Move()
    {
        Console.WriteLine($"[Animal] {Name} перемещается.");
    }

    public virtual string GetInfo()
    {
        return $"[Animal] Имя: {Name}, Возраст: {Age}, Вес: {Weight:F2} кг";
    }

    public void Eat(double amount)
    {
        Weight += amount * 0.1;
        Console.WriteLine($"[Animal] {Name} поел. Новый вес: {Weight:F2} кг");
    }

    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"[Animal] Общее количество животных: {TotalAnimals}");
    }
}

public class Dog : Animal, IPet, IHunter
{
    public string Breed { get; set; }
    public int EnergyLevel { get; set; }
    public string Owner { get; set; }

    public Dog(string name, int age, double weight, string breed, string owner)
        : base(name, age, weight)
    {
        Breed = breed;
        EnergyLevel = 100;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Dog] {Name} говорит: Гав-гав!");
    }

    public override void Move()
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} бежит. Энергия: {EnergyLevel}");
    }

    public override string GetInfo()
    {
        return $"[Dog] {Name}, {Age} лет, {Weight:F2} кг, порода: {Breed}, хозяин: {Owner}, энергия: {EnergyLevel}";
    }

    public void ShowAffection()
    {
        EnergyLevel -= 5;
        Console.WriteLine($"[Dog] {Name} виляет хвостом перед {Owner}. Энергия: {EnergyLevel}");
    }

    public void Hunt()
    {
        EnergyLevel -= 20;
        Console.WriteLine($"[Dog] {Name} охотится! Энергия: {EnergyLevel}");
    }

    public void BringSlippers()
    {
        EnergyLevel -= 10;
        Console.WriteLine($"[Dog] {Name} принёс тапочки! Энергия: {EnergyLevel}");
    }
}

public class Cat : Animal, IPet, IHunter
{
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }
    public string Owner { get; set; }

    public Cat(string name, int age, double weight, bool isIndoor, string owner)
        : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        LivesLeft = 9;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Cat] {Name} говорит: Мяу!");
    }

    public override void Move()
    {
        Console.WriteLine($"[Cat] {Name} грациозно идёт. Жизней: {LivesLeft}");
    }

    public override string GetInfo()
    {
        return $"[Cat] {Name}, {Age} лет, {Weight:F2} кг, домашний: {IsIndoor}, жизней: {LivesLeft}, хозяин: {Owner}";
    }

    public void ShowAffection()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет рядом с {Owner}.");
    }

    public void Hunt()
    {
        Console.WriteLine($"[Cat] {Name} охотится на мышей!");
    }

    public void Purr()
    {
        Console.WriteLine($"[Cat] {Name} мурлычет...");
    }
}

public class Bird : Animal, IFlyable, IPet
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    public string Owner { get; set; }

    public Bird(string name, int age, double weight, double wingspan, bool canFly)
        : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
        Owner = "Без хозяина";
    }

    public override void MakeSound()
    {
        Console.WriteLine($"[Bird] {Name} чирикает: Чик-чирик!");
    }

    public override void Move()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летит с размахом {Wingspan} м.");
        else
            Console.WriteLine($"[Bird] {Name} идёт пешком.");
    }

    public override string GetInfo()
    {
        return $"[Bird] {Name}, {Age} лет, {Weight:F2} кг, размах: {Wingspan} м, летает: {CanFly}, хозяин: {Owner}";
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"[Bird] {Name} летает!");
        else
            Console.WriteLine($"[Bird] {Name} не умеет летать.");
    }

    public void ShowAffection()
    {
        Console.WriteLine($"[Bird] {Name} сидит на плече у {Owner}.");
    }

    public void Sing()
    {
        Console.WriteLine($"[Bird] {Name} поёт!");
    }
}

// Сервис, который работает с любым IAnimal через DI
public class AnimalService
{
    private readonly IAnimal _animal;

    // Внедрение зависимости через конструктор
    public AnimalService(IAnimal animal)
    {
        _animal = animal;
    }

    public void InteractWithAnimal()
    {
        Console.WriteLine("--- AnimalService ---");
        _animal.MakeSound();
        _animal.Move();
        Console.WriteLine(_animal.GetInfo());
        Console.WriteLine();
    }
}

// Сервис для домашних животных — зависит от IPet
public class PetService
{
    private readonly IPet _pet;

    public PetService(IPet pet)
    {
        _pet = pet;
    }

    public void CareForPet()
    {
        Console.WriteLine("--- PetService ---");
        _pet.ShowAffection();
        Console.WriteLine();
    }
}

// Сервис для охотников — зависит от IHunter
public class HunterService
{
    private readonly IHunter _hunter;

    public HunterService(IHunter hunter)
    {
        _hunter = hunter;
    }

    public void SendToHunt()
    {
        Console.WriteLine("--- HunterService ---");
        _hunter.Hunt();
        Console.WriteLine();
    }
}

// Сервис для летающих — зависит от IFlyable
public class FlightService
{
    private readonly IFlyable _flyable;

    public FlightService(IFlyable flyable)
    {
        _flyable = flyable;
    }

    public void TestFlight()
    {
        Console.WriteLine("--- FlightService ---");
        _flyable.Fly();
        Console.WriteLine();
    }
}

Dog dog = new Dog("Рекс", 5, 25.0, "Лабрадор", "Иван");
Cat cat = new Cat("Мурка", 2, 4.2, true, "Мария");
Bird bird = new Bird("Кеша", 1, 0.4, 0.3, true);

Console.WriteLine("Прямое использование через интерфейсы\n");

IAnimal[] animals = { dog, cat, bird };
foreach (IAnimal a in animals)
{
    a.MakeSound();
    a.Move();
    Console.WriteLine(a.GetInfo());
    Console.WriteLine();
}

Console.WriteLine("УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ через AnimalService\n");

// Внедряем разных животных в один и тот же сервис
AnimalService dogService = new AnimalService(dog);
dogService.InteractWithAnimal();

AnimalService catService = new AnimalService(cat);
catService.InteractWithAnimal();

AnimalService birdService = new AnimalService(bird);
birdService.InteractWithAnimal();
````

````
Console.WriteLine("Управление зависимостями: PetService\n");

PetService dogPetService = new PetService(dog);
dogPetService.CareForPet();

PetService catPetService = new PetService(cat);
catPetService.CareForPet();

PetService birdPetService = new PetService(bird);
birdPetService.CareForPet();

Console.WriteLine("Управление зависимостями: HunterService\n");

HunterService dogHunterService = new HunterService(dog);
dogHunterService.SendToHunt();

HunterService catHunterService = new HunterService(cat);
catHunterService.SendToHunt();

Console.WriteLine("Управление зависимостями: FlightService\n");

FlightService birdFlightService = new FlightService(bird);
birdFlightService.TestFlight();
````

````
Console.WriteLine("Один сервис — разные зависимости\n");

IAnimal[] allAnimals = { dog, cat, bird };
foreach (IAnimal a in allAnimals)
{
    AnimalService service = new AnimalService(a);
    service.InteractWithAnimal();
}

dog.BringSlippers();
cat.Purr();
bird.Sing();
````
